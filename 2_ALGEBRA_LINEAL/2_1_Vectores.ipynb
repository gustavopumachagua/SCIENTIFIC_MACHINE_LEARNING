# 1. ¿Qué es un vector?

Un vector es un objeto matemático que puede representar:

* magnitud;
* dirección;
* desplazamiento;
* velocidad;
* fuerza;
* posición;
* características de un objeto;
* parámetros de un modelo.

En 2 dimensiones podemos escribir:

$$
\mathbf v=
\begin{bmatrix}
v_1\\
v_2
\end{bmatrix}
$$

Por ejemplo:

$$
\mathbf v=
\begin{bmatrix}
3\\
4
\end{bmatrix}
$$

Geométricamente representa una flecha que va desde:

$$
(0,0)
$$

hasta:

$$
(3,4)
$$

# 2. Vector como desplazamiento

Supongamos que estás en:

$$
A=(2,1)
$$

y quieres llegar a:

$$
B=(7,5)
$$

El desplazamiento es:

$$
\mathbf v=B-A
$$

por tanto:

$$
\mathbf v=
\begin{bmatrix}
7-2\\
5-1
\end{bmatrix}
=
\begin{bmatrix}
5\\
4
\end{bmatrix}

El vector no representa necesariamente una posición.

Representa el **cambio**.

# 3. Gráfica de un vector con Python

````python
import numpy as np
import matplotlib.pyplot as plt

A = np.array([2, 1])
B = np.array([7, 5])

v = B - A

plt.figure(figsize=(8, 8))

plt.quiver(
    A[0], A[1],
    v[0], v[1],
    angles="xy",
    scale_units="xy",
    scale=1
)

plt.scatter(*A, s=80, label="A")
plt.scatter(*B, s=80, label="B")

plt.text(A[0], A[1], "  A")
plt.text(B[0], B[1], "  B")

plt.xlim(0, 9)
plt.ylim(0, 7)

plt.axhline(0)
plt.axvline(0)

plt.xlabel("x")
plt.ylabel("y")
plt.title("Vector como desplazamiento")
plt.grid()
plt.legend()

plt.show()
````

La flecha representa:

$$
\boxed{\mathbf v=(5,4)}

# 6. Norma de un vector


La norma representa el tamaño o longitud del vector.

Para:

$$
\mathbf v=
\begin{bmatrix}
v_1\\
v_2
\end{bmatrix}
$$

la norma euclidiana es:

$$
\boxed{
\|\mathbf v\|_2=
\sqrt{v_1^2+v_2^2}
}
$$

## Ejemplo

Para:

$$
\mathbf v=
\begin{bmatrix}
3\\
4
\end{bmatrix}
$$

tenemos:

$$
\|\mathbf v\|
=
\sqrt{3^2+4^2}
$$

$$
=
\sqrt{25}
$$

$$
\boxed{5}
$$

Esto es simplemente el teorema de Pitágoras.

# 7. Norma en Python

````python
import numpy as np

v = np.array([3, 4])

norma = np.linalg.norm(v)

print(norma)
````

Resultado:


# 9. Diferentes normas

No existe solamente una norma.

## Norma $(L^2)$

$$
\boxed{
\|\mathbf x\|_2=
\sqrt{\sum_i x_i^2}
}
$$

Es la norma euclidiana.

## Norma $(L^1)$

$$
\boxed{
\|\mathbf x\|_1=
\sum_i|x_i|
}
$$

Por ejemplo:

$$
\mathbf x=(3,-4,2)
$$

entonces:

$$
\|\mathbf x\|_1=3+4+2=9
$$

## Norma infinito

$$
\boxed{
\|\mathbf x\|_\infty
=
\max_i|x_i|
}
$$

Entonces:

$$
\|(3,-4,2)\|_\infty=4
$$

# 13. Producto interno

Ahora viene uno de los conceptos más importantes.

Dados:

$$
\mathbf u=
\begin{bmatrix}

$$
\boxed{
\mathbf u\cdot\mathbf v

u_1v_1+u_2v_2

En $(n)$ dimensiones:

$$
\boxed{
\mathbf u\cdot\mathbf v
=
\sum_{i=1}^{n}u_iv_i
}
$$

# 15. La interpretación geométrica del producto interno

Aquí está la parte realmente importante:

$$
\boxed{
\mathbf u\cdot\mathbf v
=
\|\mathbf u\|
\|\mathbf v\|
\cos\theta
}
$$

donde:

$$
\theta
$$

es el ángulo entre ambos vectores.

Por tanto:

### Si:

$$
\theta<90^\circ
$$

entonces:

$$
\mathbf u\cdot\mathbf v>0
$$

### Si:


$$
\theta=90^\circ
$$

entonces:

$$
\boxed{
\mathbf u\cdot\mathbf v=0
}
$$

### Si:

$$
\theta>90^\circ
$$

entonces:

$$
\mathbf u\cdot\mathbf v<0
$$

# 19. Proyección de un vector


Este concepto es **fundamental** para SciML.

Supongamos:

$$
\mathbf u
$$

y queremos saber:

> ¿Cuánto de $(\mathbf u)$ está en la dirección de $(\mathbf v)$?

La respuesta es la **proyección**.

La proyección de $(\mathbf u)$ sobre $(\mathbf v)$ es:

$$
\boxed{
\operatorname{proj}_{\mathbf v}\mathbf u
=
\frac{\mathbf u\cdot\mathbf v}
{\mathbf v\cdot\mathbf v}
\mathbf v
}
$$

# 20. Ejemplo difícil de proyección

Tenemos:

$$
\mathbf u=
\begin{bmatrix}
5\\
3
\end{bmatrix}
$$

y:

$$
\mathbf v=
\begin{bmatrix}
4\\
1
\end{bmatrix}
$$

Primero:

$$
\mathbf u\cdot\mathbf v
=
5(4)+3(1)
=
23
$$

Luego:

$$
\mathbf v\cdot\mathbf v
=
4^2+1^2
=
17
$$

Entonces:

$$
\operatorname{proj}_{\mathbf v}\mathbf u
=
\frac{23}{17}
\begin{bmatrix}
4\\
1
\end{bmatrix}
$$

Por tanto:

$$
\boxed{
\operatorname{proj}_{\mathbf v}\mathbf u
=
\begin{bmatrix}
92/17\\
23/17
\end{bmatrix}
}
$$

aproximadamente:

$$
\boxed{
(5.412,\;1.353)
}
$$

# 21. Python para la proyección

````python
import numpy as np

u = np.array([5, 3])
v = np.array([4, 1])

projection = (
    np.dot(u, v)
    / np.dot(v, v)
) * v

print(projection)
````

# 22. Gráfica de la proyección

````python
import numpy as np
import matplotlib.pyplot as plt

u = np.array([5, 3])
v = np.array([4, 1])

projection = (
    np.dot(u, v)
    / np.dot(v, v)
) * v

plt.figure(figsize=(9, 7))

plt.quiver(
    0, 0,
    u[0], u[1],
    angles="xy",
    scale_units="xy",
    scale=1,
    label="u"
)

plt.quiver(
    0, 0,
    v[0], v[1],
    angles="xy",
    scale_units="xy",
    scale=1,
    label="v"
)

plt.quiver(
    0, 0,
    projection[0], projection[1],
    angles="xy",
    scale_units="xy",
    scale=1,
    label="Proyección"
)

plt.plot(
    [u[0], projection[0]],
    [u[1], projection[1]],
    linestyle="--",
    label="Error perpendicular"
)

plt.xlim(-1, 7)
plt.ylim(-1, 5)

plt.axhline(0)
plt.axvline(0)

plt.xlabel("x")
plt.ylabel("y")
plt.title("Proyección de u sobre v")
plt.grid()
plt.legend()

plt.show()
````

Observa una propiedad importantísima:

$$
\mathbf u-\operatorname{proj}_{\mathbf v}\mathbf u
$$

es perpendicular a:

$$
\mathbf v
$$

# 24. Bases

Ahora vamos a una idea todavía más profunda.

Una **base** es un conjunto de vectores que permite representar todos los elementos de un espacio mediante combinaciones lineales.

En $(\mathbb R^2)$, la base estándar es:

$$
\boxed{
\mathbf e_1=
\begin{bmatrix}
1\\
0
\end{bmatrix},
\qquad
\mathbf e_2=
\begin{bmatrix}
0\\
1
\end{bmatrix}
}
$$

Cualquier vector:

$$
\mathbf v=
\begin{bmatrix}
3\\
5
\end{bmatrix}
$$

puede escribirse:

$$
\mathbf v
=
3\mathbf e_1+5\mathbf e_2
$$

# 25. Base no estándar

Consideremos:

$$
\mathbf b_1=
\begin{bmatrix}
1\\
1
\end{bmatrix}
$$

y:

$$
\mathbf b_2=
\begin{bmatrix}
1\\
-1
\end{bmatrix}
$$

También pueden formar una base de $(\mathbb R^2)$.

Queremos representar:

$$
\mathbf v=
\begin{bmatrix}
5\\
1
\end{bmatrix}
$$

como:

$$
\mathbf v
=
c_1\mathbf b_1+c_2\mathbf b_2
$$

Esto significa:

$$
c_1
\begin{bmatrix}
1\\
1
\end{bmatrix}
+
c_2
\begin{bmatrix}
1\\
-1
\end{bmatrix}
=
\begin{bmatrix}
5\\
1
\end{bmatrix}
$$

Por componentes:

$$
c_1+c_2=5
$$

$$
c_1-c_2=1
$$

Sumando:

$$
2c_1=6
$$

$$
c_1=3
$$

Entonces:

$$
c_2=2
$$

Así:

$$
\boxed{
\mathbf v=3\mathbf b_1+2\mathbf b_2
}
$$

# 30. De una PDE a un vector

Tenemos:

$$
u(x,t)
$$

y discretizamos:

$$
x_1,x_2,\ldots,x_N
$$

Entonces:

$$
u(x_1,t)\rightarrow u_1(t)
$$

$$
u(x_2,t)\rightarrow u_2(t)
$$

etc.

Así obtenemos:

$$
\boxed{
\mathbf u(t)
=
[u_1(t),u_2(t),\ldots,u_N(t)]^T
}
$$

Una PDE puede terminar convertida en:

$$
\boxed{
\frac{d\mathbf u}{dt}
=
F(\mathbf u,t)
}
$$

¡Y ahora tenemos un sistema de ecuaciones diferenciales ordinarias!

Esto conecta:

$$
\boxed{
\text{Cálculo}
\rightarrow
\text{vectores}
\rightarrow
\text{PDE}
\rightarrow
\text{SciML}
}
$$

Por tanto, estos tres vectores **no forman una base de $(\mathbb R^3)$**.

# 32. Independencia lineal

Los vectores:

$$
v_1,\ldots,v_n
$$

son linealmente independientes si:

$$
c_1v_1+\cdots+c_nv_n=0
$$

implica necesariamente:

$$
\boxed{
c_1=c_2=\cdots=c_n=0
}
$$

Esto es esencial para entender bases.

# 33. Una conexión profunda: producto interno → ortogonalidad → proyección → base

Observa la cadena:

$$

}
$$

nos permite determinar el ángulo.

Si:

$$

$$

tenemos:

\boxed{\text{ortogonalidad}}
$$

La ortogonalidad nos permite construir bases especialmente convenientes:

$$
\boxed{\text{bases ortogonales}}

Esto conduce a:

# 34. Gram-Schmidt

Supongamos que tenemos:

$$
v_1,v_2
$$

Podemos convertirlos en vectores ortogonales.

Primero:

$$
u_1=v_1
$$

Luego:

$$
u_2=
v_2-
\operatorname{proj}_{u_1}(v_2)
$$

Así:

$$
u_1\perp u_2
$$

Después normalizamos:

$$
e_1=
\frac{u_1}{\|u_1\|}
$$

$$
e_2=
\frac{u_2}{\|u_2\|}
$$

y obtenemos una base ortonormal.

Esto será importantísimo para:

* QR;
* mínimos cuadrados;
* PCA;
* métodos numéricos;
* métodos espectrales;
* procesamiento de señales;
* ML.

# 35. Ejemplo Python completo

````python
import numpy as np

v1 = np.array([1.0, 1.0])
v2 = np.array([1.0, 2.0])

# Primer vector
u1 = v1

# Proyección de v2 sobre u1
projection = (
    np.dot(v2, u1)
    / np.dot(u1, u1)
) * u1

# Segundo vector ortogonal
u2 = v2 - projection

# Normalización
e1 = u1 / np.linalg.norm(u1)
e2 = u2 / np.linalg.norm(u2)

print("e1 =", e1)
print("e2 =", e2)

print("Producto interno:")
print(np.dot(e1, e2))
````

El último resultado debería ser aproximadamente:

```text
0
```

por errores numéricos de punto flotante puede aparecer algo como:

```text
2.22e-16
```

Eso numéricamente significa prácticamente cero.

# 36. Un ejemplo más cercano a Machine Learning: distancia entre datos

Supongamos dos observaciones:

$$
x_1=
\begin{bmatrix}
1\\
70\\
180
\end{bmatrix}
$$

y:

$$
x_2=
\begin{bmatrix}
2\\
75\\
175
\end{bmatrix}
$$

Podríamos calcular:

$$
d(x_1,x_2)
$$

Pero hay un problema.

Las variables tienen escalas diferentes.

Por ejemplo:

* edad ≈ 1–100;
* peso ≈ 50–150;
* altura ≈ 150–200.

La distancia euclidiana puede estar dominada por las variables de mayor escala.

Esto lleva a una práctica fundamental de Machine Learning:

$$
\boxed{\text{normalización / estandarización}}
$$

Antes de utilizar distancias en muchos algoritmos.

# 37. Distancia en espacios de alta dimensión

En Machine Learning aparecen vectores como:

$$
x\in\mathbb R^{100}
$$

o:

$$
x\in\mathbb R^{10000}
$$

La distancia:

$$
d(x,y)=\|x-y\|_2
$$

sigue funcionando.

Pero aparece un fenómeno conocido como:

$$
\boxed{\text{curse of dimensionality}}
$$

En dimensiones altas, algunas intuiciones geométricas de 2D/3D dejan de comportarse como esperaríamos.

Esto será muy importante posteriormente en ML.

# 38. Un ejemplo de error científico

Supongamos:

$$
u_{\text{real}}
=
\begin{bmatrix}
1.0\\
2.0\\
3.0\\
4.0
\end{bmatrix}
$$

y nuestro modelo predice:

$$
u_{\text{pred}}
=
\begin{bmatrix}
1.1\\
1.9\\
3.2\\
3.8
\end{bmatrix}
$$

El error es:

$$
e=u_{\text{real}}-u_{\text{pred}}
$$

$$
e=
\begin{bmatrix}
-0.1\\
0.1\\
-0.2\\
0.2
\end{bmatrix}
$$

Podemos medir:

$$
\|e\|_2
$$

con:

````python
u_real = np.array([1, 2, 3, 4])

u_pred = np.array([
    1.1,
    1.9,
    3.2,
    3.8
])

error = u_real - u_pred

print("Error vector:", error)
print("Norma L2:", np.linalg.norm(error))
print("Norma L1:", np.linalg.norm(error, 1))
print("Norma infinito:", np.linalg.norm(error, np.inf))
````

Esto es mucho más cercano a lo que harás en SciML que los ejemplos puramente geométricos.
