# 1. La idea central: ¿qué es un espacio vectorial?

Antes de hablar de subespacios, bases, etc., necesitamos entender qué estamos estudiando.

Un **espacio vectorial** es un conjunto de objetos que podemos:

1. sumar;
2. multiplicar por escalares;
3. y esas operaciones cumplen ciertas reglas.

Por ejemplo:

$$
\mathbb R^2=
\left\{
\begin{bmatrix}
x\\y
\end{bmatrix}:x,y\in\mathbb R
\right\}
$$

es un espacio vectorial.

También:

$$
\mathbb R^3
$$

y, muy importante para SciML:

$$
\mathbb R^N.
$$

Pero también existen espacios mucho más interesantes:

$$
P_n=\{\text{polinomios de grado }\le n\}
$$

y espacios de funciones:

$$
C[a,b]=\{f:[a,b]\rightarrow\mathbb R
\mid f\text{ continua}\}.
$$

Esto último es importantísimo en SciML porque las soluciones de EDO/PDE son **funciones**, no simplemente números.

Por ejemplo, si tenemos la ecuación de calor

$$
\frac{\partial u}{\partial t}
=
\alpha\frac{\partial^2u}{\partial x^2},
$$

la solución

$$
u(x,t)
$$

es una función.

Por eso eventualmente pasarás de

$$
\mathbf x\in\mathbb R^N
$$

a

$$
u(x,t)\in\mathcal V,
$$

donde $(\mathcal V)$ es un espacio de funciones.

# 2. SPAN — conjunto generado

Este es uno de los conceptos más importantes.

Supongamos que tenemos

$$
v_1=
\begin{bmatrix}
1\\2
\end{bmatrix},
\qquad
v_2=
\begin{bmatrix}
3\\1
\end{bmatrix}.
$$

El **span** de estos vectores es el conjunto de todas las combinaciones lineales posibles:

$$
\operatorname{span}(v_1,v_2)
=
\left\{
c_1v_1+c_2v_2:
c_1,c_2\in\mathbb R
\right\}.
$$

Es decir:

$$
c_1
\begin{bmatrix}
1\\2
\end{bmatrix}
+
c_2
\begin{bmatrix}
3\\1
\end{bmatrix}.
$$

Entonces:

$$
=
\begin{bmatrix}
c_1+3c_2\\
2c_1+c_2
\end{bmatrix}.
$$

La pregunta importante es:

> ¿Qué parte del espacio $(\mathbb R^2)$ podemos construir utilizando estos vectores?

Como los vectores no son paralelos, podemos generar todo $(\mathbb R^2)$:

$$
\boxed{
\operatorname{span}(v_1,v_2)=\mathbb R^2
}
$$

### Visualmente

Los dos vectores generan un paralelogramo fundamental, pero permitiendo todos los coeficientes reales podemos llegar a cualquier punto del plano.

## Ejemplo más difícil

Considera:

$$
v_1=
\begin{bmatrix}
1\\
2\\
3
\end{bmatrix},
\quad
v_2=
\begin{bmatrix}
2\\
4\\
6
\end{bmatrix},
\quad
v_3=
\begin{bmatrix}
1\\
0\\
1
\end{bmatrix}.
$$

Observa que:

$$
v_2=2v_1.
$$

Por lo tanto, $(v_2)$ no aporta una nueva dirección.

El span realmente es:

$$
\operatorname{span}(v_1,v_2,v_3)
=
\operatorname{span}(v_1,v_3).
$$

¿Generan todo $(\mathbb R^3)$?

Tenemos solamente dos direcciones independientes.

Por tanto:

$$
\boxed{
\dim(\operatorname{span})=2
}
$$

y geométricamente generan un **plano dentro de $(\mathbb R^3)$**.

# 3. ¿Para qué sirve el span en SciML?

Aquí comienza a ponerse interesante.

Supongamos que una solución de una PDE complicada es:

$$
u(x,t).
$$

Queremos aproximarla utilizando unas funciones conocidas:

$$
\phi_1(x),\phi_2(x),\ldots,\phi_N(x).
$$

Construimos:

$$
u(x,t)
\approx
c_1(t)\phi_1(x)
+
c_2(t)\phi_2(x)
+\cdots+
c_N(t)\phi_N(x).
$$

Esto significa:

$$
u(x,t)\in
\operatorname{span}
\{\phi_1,\ldots,\phi_N\}.
$$

**Eso es exactamente un span.**

Por ejemplo:

$$
u(x)
\approx
c_1\sin(x)
+c_2\sin(2x)
+c_3\sin(3x).
$$

Entonces estamos diciendo:

$$
u\in
\operatorname{span}
\{\sin x,\sin2x,\sin3x\}.
$$

Esto aparece en:

* métodos espectrales;
* Fourier;
* elementos finitos;
* reducción de modelos;
* Galerkin;
* PINNs;
* aproximación numérica de PDEs.

# 4. SUBESPACIOS

Un **subespacio** es un subconjunto de un espacio vectorial que, por sí mismo, sigue siendo un espacio vectorial.

Supongamos:

$$
V=\mathbb R^3.
$$

Consideremos:

$$
W=
\left\{
\begin{bmatrix}
x\\y\\z
\end{bmatrix}
:
x+2y-z=0
\right\}.
$$

¿Es $(W)$ un subespacio?

Tenemos que verificar tres cosas.

## Propiedad 1 — contiene al vector cero

Ponemos:

$$
x=y=z=0.
$$

Entonces:

$$
0+2(0)-0=0.
$$

Sí:

$$
\mathbf 0\in W.
$$

## Propiedad 2 — cerrado bajo suma

Supongamos:

$$
u,v\in W.
$$

Entonces:

$$
u_1+2u_2-u_3=0
$$

y

$$
v_1+2v_2-v_3=0.
$$

Para:

$$
u+v
$$

tenemos:

$$
(u_1+v_1)
+2(u_2+v_2)
-(u_3+v_3).
$$

Agrupamos:

$$
=
(u_1+2u_2-u_3)
+
(v_1+2v_2-v_3)
=0.
$$

Por tanto:

$$
u+v\in W.
$$

## Propiedad 3 — cerrado bajo multiplicación escalar

Sea:

$$
\lambda u.
$$

Entonces:

$$
(\lambda u_1)+2(\lambda u_2)-\lambda u_3
$$

$$
=
\lambda(u_1+2u_2-u_3)
=0.
$$

Por tanto:

$$
\lambda u\in W.
$$

Así que:

$$
\boxed{W\text{ es un subespacio de }\mathbb R^3}
$$

# 5. Geometría del subespacio

La ecuación:

$$
x+2y-z=0
$$

puede escribirse:

$$
z=x+2y.
$$

Es un plano que pasa por el origen.

Eso es muy importante:

### Un subespacio de $(\mathbb R^3)$ puede ser:

* $(\{0\})$;
* una recta que pasa por el origen;
* un plano que pasa por el origen;
* todo $(\mathbb R^3)$.

Pero un plano que **no pasa por el origen** no es subespacio.

Por ejemplo:

$$
x+y+z=1
$$

NO es subespacio porque:

$$
(0,0,0)
$$

no pertenece al conjunto.

# 6. Ejemplo aplicado: estados físicos

Imagina que tenemos tres variables:

$$
\mathbf x=
\begin{bmatrix}
T_1\\
T_2\\
T_3
\end{bmatrix}
$$

representando temperaturas de tres puntos de una placa.

Supongamos que físicamente imponemos:

$$
T_1+2T_2-T_3=0.
$$

Entonces los estados físicamente permitidos pertenecen al subespacio:

$$
W=
\{\mathbf x:A\mathbf x=0\}.
$$

Esto nos lleva directamente al **espacio nulo**, que veremos después.

# 7. INDEPENDENCIA LINEAL

Ahora viene uno de los conceptos más importantes.

Tenemos vectores:

$$
v_1,v_2,\ldots,v_n.
$$

Son linealmente independientes si la única manera de obtener cero es:

$$
c_1v_1+\cdots+c_nv_n=0
$$

con

$$
\boxed{c_1=c_2=\cdots=c_n=0}.
$$

Si existe alguna combinación no trivial:

$$
(c_1,\ldots,c_n)\neq0
$$

que produzca:

$$
0,
$$

entonces son **dependientes**.

# 8. Ejemplo difícil de independencia

Consideremos:

$$
v_1=
\begin{bmatrix}
1\\2\\0\\1
\end{bmatrix},
\quad
v_2=
\begin{bmatrix}
0\\1\\1\\2
\end{bmatrix},
\quad
v_3=
\begin{bmatrix}
2\\5\\1\\4
\end{bmatrix}.
$$

Queremos saber si son independientes.

Planteamos:

$$
c_1v_1+c_2v_2+c_3v_3=0.
$$

Esto produce:

$$
c_1+2c_3=0
$$

$$
2c_1+c_2+5c_3=0
$$

$$
c_2+c_3=0
$$

$$
c_1+2c_2+4c_3=0.
$$

De la primera:

$$
c_1=-2c_3.
$$

De la tercera:

$$
c_2=-c_3.
$$

Sustituimos en la segunda:

$$
2(-2c_3)-c_3+5c_3=0
$$

$$
-4c_3-c_3+5c_3=0.
$$

Esto no nos da una restricción.

Probamos la cuarta:

$$
-2c_3+2(-c_3)+4c_3=0.
$$

También se cumple.

Por tanto podemos escoger:

$$
c_3=1.
$$

Entonces:

$$
c_1=-2,
\qquad
c_2=-1,
\qquad
c_3=1.
$$

Por tanto:

$$
-2v_1-v_2+v_3=0.
$$

Así:

$$
\boxed{
v_3=2v_1+v_2
}
$$

y los vectores son **linealmente dependientes**.

# 9. Una forma geométrica de entender independencia

En $(\mathbb R^2)$:

Dos vectores son independientes si no están sobre la misma recta.

En $(\mathbb R^3)$:

Tres vectores son independientes si no están todos contenidos en un mismo plano que pase por el origen.

En general:

$$
n\text{ vectores independientes}
$$

significa que cada vector aporta una **nueva dirección**.

Esta idea será fundamental cuando estudiemos:

* bases;
* eigenvectors;
* SVD;
* PCA;
* reducción dimensional;
* métodos numéricos.

# 10. BASES

Una base es un conjunto de vectores que cumple dos condiciones:

### 1. Generan el espacio


$$
\operatorname{span}\{v_1,\ldots,v_n\}=V.
$$

### 2. Son linealmente independientes

Por tanto:

$$
\boxed{
\text{BASE}
=
\text{SPAN}
+
\text{INDEPENDENCIA}
}
$$

Una base de $(\mathbb R^2)$ puede ser:

$$
e_1=
\begin{bmatrix}
1\\0
\end{bmatrix},
\qquad
e_2=
\begin{bmatrix}
0\\1
\end{bmatrix}.
$$

La base estándar.

Pero también:

$$
b_1=
\begin{bmatrix}
1\\1
\end{bmatrix},
\qquad
b_2=
\begin{bmatrix}
1\\-1
\end{bmatrix}.
$$

También es una base.

# 11. ¿Por qué una base es tan importante?

Porque nos permite representar cualquier vector mediante coordenadas.

Por ejemplo:

$$
v=
\begin{bmatrix}
5\\1
\end{bmatrix}.
$$

Usando la base:

$$
b_1=
\begin{bmatrix}
1\\1
\end{bmatrix},
\qquad
b_2=
\begin{bmatrix}
1\\-1
\end{bmatrix},
$$

queremos:

$$
v=c_1b_1+c_2b_2.
$$

Entonces:

$$
c_1
\begin{bmatrix}
1\\1
\end{bmatrix}
+
c_2
\begin{bmatrix}
1\\-1
\end{bmatrix}
=
\begin{bmatrix}
5\\1
\end{bmatrix}.
$$

Tenemos:

$$
c_1+c_2=5
$$

$$
c_1-c_2=1.
$$

Sumamos:

$$
2c_1=6
$$

$$
c_1=3.
$$

Entonces:

$$
c_2=2.
$$

Por tanto:

$$
\boxed{
[v]_B=
\begin{bmatrix}
3\\2
\end{bmatrix}
}
$$

Aunque en coordenadas estándar:

$$
[v]_E=
\begin{bmatrix}
5\\1
\end{bmatrix}.
$$

El mismo vector tiene coordenadas diferentes dependiendo de la base.

# 12. Aplicación científica: bases de funciones

Esto es muchísimo más importante para SciML.

Considera:

$$
u(x)=3\sin x-2\cos x+5\sin(2x).
$$

Podemos definir la base:

$$
B=
\{\sin x,\cos x,\sin(2x)\}.
$$

Entonces:

$$
[u]_B=
\begin{bmatrix}
3\\
-2\\
5
\end{bmatrix}.
$$

Esto es precisamente la idea detrás de:

* Fourier;
* métodos espectrales;
* elementos finitos;
* funciones de base radial;
* aproximaciones de PDE.

Una función complicada puede convertirse en un vector de coeficientes.

# 13. DIMENSIÓN

La dimensión es el número de vectores de una base.

Por ejemplo:

$$
\dim(\mathbb R^2)=2
$$

$$
\dim(\mathbb R^3)=3.
$$

Pero cuidado:

Un plano dentro de $(\mathbb R^3)$ tiene:

$$
\boxed{\dim(W)=2}
$$

aunque sus vectores tengan tres componentes.

Por ejemplo:

$$
W=
\left\{
(x,y,z):
x+2y-z=0
\right\}.
$$

Podemos escribir:

$$
z=x+2y.
$$

Entonces:

$$
\begin{bmatrix}
x\\y\\z
\end{bmatrix}
=
\begin{bmatrix}
x\\y\\x+2y
\end{bmatrix}
$$

$$
=
x
\begin{bmatrix}
1\\0\\1
\end{bmatrix}
+
y
\begin{bmatrix}
0\\1\\2
\end{bmatrix}.
$$

Por tanto:

$$
W=
\operatorname{span}
\left\{
\begin{bmatrix}
1\\0\\1
\end{bmatrix},
\begin{bmatrix}
0\\1\\2
\end{bmatrix}
\right\}.
$$

Una base es:

$$
B=
\left\{
\begin{bmatrix}
1\\0\\1
\end{bmatrix},
\begin{bmatrix}
0\\1\\2
\end{bmatrix}
\right\}.
$$

Por tanto:

$$
\boxed{\dim W=2}.
$$

# 14. ESPACIO COLUMNA

Ahora entramos en una conexión fundamental con matrices.

Sea:

$$
A=
\begin{bmatrix}
1&2&4\\
0&1&3\\
2&5&11
\end{bmatrix}.
$$

Las columnas son:

$$
a_1=
\begin{bmatrix}
1\\0\\2
\end{bmatrix},
\quad
a_2=
\begin{bmatrix}
2\\1\\5
\end{bmatrix},
\quad
a_3=
\begin{bmatrix}
4\\3\\11
\end{bmatrix}.
$$

El **espacio columna** es:

$$
\boxed{
\operatorname{Col}(A)
=
\operatorname{span}\{a_1,a_2,a_3\}
}
$$

Es decir, todos los vectores que podemos producir mediante:

$$
A x.
$$

Porque:

$$
A
\begin{bmatrix}
x_1\\x_2\\x_3
\end{bmatrix}
=
x_1a_1+x_2a_2+x_3a_3.
$$

Por tanto:

$$
\boxed{
\operatorname{Col}(A)=\{Ax:x\in\mathbb R^n\}
}
$$

# 15. La conexión con sistemas lineales

Cuando resolvemos:

$$
Ax=b,
$$

la pregunta realmente es:

> ¿$(b)$ pertenece al espacio columna de $(A)$?

Si:

$$
b\in\operatorname{Col}(A),
$$

el sistema tiene al menos una solución.

Si:

$$
b\notin\operatorname{Col}(A),
$$

no existe solución exacta.

Esta interpretación es extremadamente importante en:

* ecuaciones diferenciales discretizadas;
* regresión;
* mínimos cuadrados;
* problemas inversos;
* física computacional.

# 16. ESPACIO NULO

Ahora consideremos:

$$
Ax=0.
$$

El conjunto de todas las soluciones es el **espacio nulo**:

$$
\boxed{
\operatorname{Null}(A)
=
\{x:Ax=0\}
}
$$

Ejemplo:

$$
A=
\begin{bmatrix}
1&2&3\\
2&4&6
\end{bmatrix}.
$$

Queremos:

$$
A
\begin{bmatrix}
x\\y\\z
\end{bmatrix}
=
\begin{bmatrix}
0\\0
\end{bmatrix}.
$$

Tenemos:

$$
x+2y+3z=0.
$$

La segunda ecuación es simplemente:

$$
2(x+2y+3z)=0.
$$

Por tanto solamente tenemos una ecuación independiente.

Despejamos:

$$
x=-2y-3z.
$$

Entonces:

$$
\begin{bmatrix}
x\\y\\z
\end{bmatrix}
=
\begin{bmatrix}
-2y-3z\\
y\\
z
\end{bmatrix}.
$$

Separando:

$$
=
y
\begin{bmatrix}
-2\\1\\0
\end{bmatrix}
+
z
\begin{bmatrix}
-3\\0\\1
\end{bmatrix}.
$$

Por tanto:

$$
\boxed{
\operatorname{Null}(A)
=
\operatorname{span}
\left\{
\begin{bmatrix}
-2\\1\\0
\end{bmatrix},
\begin{bmatrix}
-3\\0\\1
\end{bmatrix}
\right\}
}
$$

y:

$$
\boxed{\dim\operatorname{Null}(A)=2}.
$$

# 17. Interpretación física del espacio nulo

Imagina que:

$$
A:\text{configuración física}\rightarrow\text{medición}.
$$

Entonces:

$$
Ax=0
$$

significa:

> configuraciones diferentes que producen exactamente cero en las mediciones.

En un problema de sensores:

* $(x)$ = estado real de un sistema;
* $(A)$ = matriz de medición;
* $(Ax)$ = datos observados.

Si existe:

$$
x\neq0
$$

pero

$$
Ax=0,
$$

entonces existe una perturbación del estado que los sensores **no pueden detectar**.

Eso es una idea central en problemas inversos y SciML.

# 18. Teorema rango-nulidad

Uno de los teoremas más importantes:

Si:

$$
A\in\mathbb R^{m\times n},
$$

entonces:

$$
\boxed{
\operatorname{rank}(A)
+
\operatorname{nullity}(A)
=
n
}
$$

donde:

$$
\operatorname{rank}(A)
=
\dim\operatorname{Col}(A)
$$

y

$$
\operatorname{nullity}(A)
=
\dim\operatorname{Null}(A).
$$

Para la matriz anterior:

$$
A\in\mathbb R^{2\times3}.
$$

Tenemos:

$$
\operatorname{rank}(A)=1
$$

y:

$$
\operatorname{nullity}(A)=2.
$$

Entonces:

$$
1+2=3.
$$

Perfecto.

# 19. TRANSFORMACIONES LINEALES

Ahora llegamos al concepto que conecta prácticamente todo.

Una transformación:

$$
T:V\rightarrow W
$$

es lineal si cumple:

### Aditividad

$$
T(u+v)=T(u)+T(v).
$$

### Homogeneidad

$$
T(\alpha v)=\alpha T(v).
$$

Combinando ambas:

$$
\boxed{
T(\alpha u+\beta v)
=
\alpha T(u)+\beta T(v)
}
$$

# 20. Ejemplo: transformación matricial

Toda matriz define una transformación lineal:

$$
T(x)=Ax.
$$

Por ejemplo:

$$
A=
\begin{bmatrix}
2&1\\
1&3
\end{bmatrix}.
$$

Entonces:

$$
T
\begin{bmatrix}
x\\y
\end{bmatrix}
=
\begin{bmatrix}
2x+y\\
x+3y
\end{bmatrix}.
$$

La matriz transforma puntos del plano.

Podemos visualizarlo con Python:

```python
import numpy as np
import matplotlib.pyplot as plt

A = np.array([
    [2, 1],
    [1, 3]
])

# Malla de puntos
x = np.linspace(-2, 2, 9)
y = np.linspace(-2, 2, 9)

X, Y = np.meshgrid(x, y)

points = np.vstack([
    X.ravel(),
    Y.ravel()
])

transformed = A @ points

fig, ax = plt.subplots(figsize=(8, 8))

ax.scatter(
    points[0],
    points[1],
    label="Original"
)

ax.scatter(
    transformed[0],
    transformed[1],
    label="Transformado"
)

ax.axhline(0, linewidth=0.8)
ax.axvline(0, linewidth=0.8)

ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_title("Transformación lineal T(x)=Ax")
ax.legend()
ax.grid(True)

plt.show()
```

Aquí estás viendo cómo la transformación mueve toda una colección de puntos.

# 21. Transformación lineal como cambio geométrico

Una matriz puede:

* rotar;
* escalar;
* reflejar;
* comprimir;
* deformar;
* proyectar.

Por ejemplo, una rotación:

$$
A=
\begin{bmatrix}
\cos\theta&-\sin\theta\\
\sin\theta&\cos\theta
\end{bmatrix}.
$$

Código:

```python
import numpy as np
import matplotlib.pyplot as plt

theta = np.pi / 4

A = np.array([
    [np.cos(theta), -np.sin(theta)],
    [np.sin(theta),  np.cos(theta)]
])

v = np.array([2, 1])

w = A @ v

fig, ax = plt.subplots(figsize=(7, 7))

ax.quiver(
    0, 0,
    v[0], v[1],
    angles="xy",
    scale_units="xy",
    scale=1,
    label="v"
)

ax.quiver(
    0, 0,
    w[0], w[1],
    angles="xy",
    scale_units="xy",
    scale=1,
    label="T(v)"
)

ax.set_xlim(-3, 3)
ax.set_ylim(-3, 3)

ax.set_aspect("equal")
ax.grid(True)
ax.legend()

plt.show()
```

# 22. Algo MUY importante: núcleo de una transformación

Para una transformación:

$$
T:V\rightarrow W,
$$

definimos:

$$
\boxed{
\ker(T)=\{v\in V:T(v)=0\}
}
$$

Esto es exactamente el espacio nulo cuando:

$$
T(x)=Ax.
$$

Entonces:

$$
\boxed{
\ker(T)=\operatorname{Null}(A)
}
$$

El núcleo nos dice:

> ¿Qué información desaparece durante la transformación?

# 23. Imagen de una transformación


También tenemos:

$$
\boxed{
\operatorname{Im}(T)
=
\{T(v):v\in V\}
}
$$

Si:

$$
T(x)=Ax,
$$

entonces:

$$
\boxed{
\operatorname{Im}(T)=\operatorname{Col}(A)
}
$$

Esto conecta absolutamente todos los conceptos:

$$
\boxed{
\text{Transformación}
\leftrightarrow
\text{Matriz}
}
$$

$$
\boxed{
\operatorname{Im}(T)
=
\operatorname{Col}(A)
}
$$

$$
\boxed{
\ker(T)
=
\operatorname{Null}(A)
}
$$

# 24. Ejemplo grande: una transformación que pierde información

Considera:

$$
A=
\begin{bmatrix}
1&2&3\\
2&4&6
\end{bmatrix}.
$$

Tenemos:

$$
T:\mathbb R^3\rightarrow\mathbb R^2.
$$

Pero:

$$
\operatorname{rank}(A)=1.
$$

Aunque la entrada tenga tres dimensiones, la salida efectiva tiene solamente una dimensión.

¿Por qué?

Las columnas son:

$$
\begin{bmatrix}1\\2\end{bmatrix},
\quad
\begin{bmatrix}2\\4\end{bmatrix},
\quad
\begin{bmatrix}3\\6\end{bmatrix}.
$$

Todas están sobre la misma recta.

Entonces:

$$
\dim\operatorname{Im}(T)=1.
$$

Mientras que:

$$
\dim\ker(T)=2.
$$

Y:

$$
1+2=3.
$$

# 26. Conexión con Scientific Machine Learning

Aquí está lo verdaderamente importante para tu objetivo.

En SciML normalmente tendrás algo parecido a:

$$
\frac{du}{dt}=F(u,t).
$$

Después discretizas el espacio.

Por ejemplo:

$$
u(x,t)
\rightarrow
\mathbf u(t)
=
\begin{bmatrix}
u(x_1,t)\\
u(x_2,t)\\
\vdots\\
u(x_N,t)
\end{bmatrix}.
$$

Entonces:

$$
\mathbf u(t)\in\mathbb R^N.
$$

Ahora aparecen inmediatamente los espacios vectoriales.

## Aproximación mediante una base

Puedes escribir:

$$
u(x,t)
\approx
\sum_{i=1}^{r}c_i(t)\phi_i(x).
$$

En forma matricial:

$$
\mathbf u(t)\approx \Phi\mathbf c(t).
$$

Donde:

$$
\Phi=
\begin{bmatrix}
\phi_1(x_1)&\cdots&\phi_r(x_1)\\
\phi_1(x_2)&\cdots&\phi_r(x_2)\\
\vdots&&\vdots\\
\phi_1(x_N)&\cdots&\phi_r(x_N)
\end{bmatrix}.
$$

Aquí:

$$
\operatorname{Col}(\Phi)
$$

es el espacio de soluciones que podemos representar.

Si:

$$
r\ll N,
$$

hemos reducido enormemente la dimensión.

Esto se llama **reduced-order modeling**.

# 27. Ejemplo: PDE → espacio vectorial

Supongamos una temperatura:

$$
u(x,t).
$$

Podemos aproximarla:

$$
u(x,t)
\approx
c_1(t)\sin(\pi x)
+
c_2(t)\sin(2\pi x)
+
c_3(t)\sin(3\pi x).
$$

Entonces la solución vive aproximadamente en:

$$
V_r=
\operatorname{span}
\{
\sin(\pi x),
\sin(2\pi x),
\sin(3\pi x)
\}.
$$

La dimensión es:

$$
\boxed{\dim V_r=3}
$$

aunque la función original vive en un espacio infinito-dimensional.

Este cambio:

$$
\boxed{
\text{infinito-dimensional}
\rightarrow
\text{r-dimensional}
}
$$

es una de las ideas centrales de los métodos numéricos y SciML.

# 28. Python: visualizar una aproximación mediante una base

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 1, 1000)

phi1 = np.sin(np.pi * x)
phi2 = np.sin(2 * np.pi * x)
phi3 = np.sin(3 * np.pi * x)

c1 = 1.5
c2 = -0.8
c3 = 0.4

u = c1 * phi1 + c2 * phi2 + c3 * phi3

plt.figure(figsize=(10, 5))

plt.plot(x, phi1, label=r"$\phi_1=\sin(\pi x)$")
plt.plot(x, phi2, label=r"$\phi_2=\sin(2\pi x)$")
plt.plot(x, phi3, label=r"$\phi_3=\sin(3\pi x)$")

plt.plot(
    x,
    u,
    linewidth=3,
    label=r"$u=1.5\phi_1-0.8\phi_2+0.4\phi_3$"
)

plt.xlabel("x")
plt.ylabel("u(x)")
plt.title("Construcción de una función mediante una base")
plt.grid(True)
plt.legend()

plt.show()
```

Aquí estás viendo literalmente:

$$
u=c_1\phi_1+c_2\phi_2+c_3\phi_3.
$$

# 29. El mapa conceptual que quiero que tengas en la cabeza

### SPAN

\operatorname{span}\{v_1,\ldots,v_n\}

↓

### SUBESPACIO

> ¿Ese conjunto generado forma un espacio vectorial?

↓

c_1v_1+\cdots+c_nv_n=0
$$

↓

### BASE

> ¿Tengo suficientes vectores para generar el espacio y ninguno es redundante?

$$
\boxed{\text{base}=\text{generación}+\text{independencia}}
$$

↓

### DIMENSIÓN

> ¿Cuántas direcciones independientes tiene el espacio?

↓

### ESPACIO COLUMNA

> ¿Qué salidas puede producir \(Ax\)?

$$
\operatorname{Col}(A)
$$

↓

### ESPACIO NULO

> ¿Qué entradas desaparecen bajo $(A)$?

$$
\operatorname{Null}(A)
$$

↓

### TRANSFORMACIÓN LINEAL

> ¿Cómo transforma un espacio en otro?

T:V\rightarrow W
