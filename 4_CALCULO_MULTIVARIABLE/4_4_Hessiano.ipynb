# 1. Antes del Hessiano: recuerda el gradiente

Supongamos:

$$
f(x,y)
$$

El gradiente es:

$$
\boxed{
\nabla f=
\begin{bmatrix}
f_x\\
f_y
\end{bmatrix}
}
$$

El gradiente nos dice la dirección de máximo crecimiento.

Por ejemplo:

$$
f(x,y)=x^2+3y^2
$$

Entonces:

$$
\nabla f=
\begin{bmatrix}
2x\\
6y
\end{bmatrix}.
$$

Pero existe una pregunta más profunda:

> ¿Cómo cambia el gradiente cuando nos movemos?

La respuesta es el **Hessiano**.

# 2. ¿Qué es el Hessiano?

El Hessiano es la matriz formada por las **segundas derivadas parciales**.

Para:

$$
f(x,y)
$$

tenemos:

$$
\boxed{
H_f(x,y)
=
\nabla^2f
=
\begin{bmatrix}
f_{xx}&f_{xy}\\
f_{yx}&f_{yy}
\end{bmatrix}
}
$$

donde:

$$
f_{xx}=
\frac{\partial^2f}{\partial x^2}
$$

$$
f_{xy}=
\frac{\partial^2f}{\partial x\partial y}
$$

$$
f_{yx}=
\frac{\partial^2f}{\partial y\partial x}
$$

$$
f_{yy}=
\frac{\partial^2f}{\partial y^2}.
$$

Si la función es suficientemente suave:

$$
\boxed{
f_{xy}=f_{yx}
}
$$

por lo que normalmente:

$$
H=
\begin{bmatrix}
f_{xx}&f_{xy}\\
f_{xy}&f_{yy}
\end{bmatrix}.
$$

# 3. ¿Qué significa geométricamente?

Piensa en una montaña.

El gradiente te dice:

> "¿Hacia dónde está subiendo más?"

El Hessiano te dice:

> "¿Cómo está cambiando la pendiente?"

Es decir, el Hessiano mide la **curvatura local**.

Dos superficies pueden tener exactamente el mismo gradiente en un punto, pero tener curvaturas completamente diferentes.

# 4. Ejemplo básico

Consideremos:

$$
f(x,y)=x^2+3y^2
$$

Primero:

$$
f_x=2x
$$

$$
f_y=6y
$$

Entonces:

$$
\nabla f=
\begin{bmatrix}
2x\\
6y
\end{bmatrix}.
$$

Ahora derivamos nuevamente.

$$
f_{xx}=2
$$

$$
f_{xy}=0
$$

$$
f_{yx}=0
$$

$$
f_{yy}=6.
$$

Por tanto:

$$
\boxed{
H=
\begin{bmatrix}
2&0\\
0&6
\end{bmatrix}
}
$$

Este Hessiano nos dice que la superficie tiene más curvatura en la dirección $(y)$ que en $(x)$.

# 6. Hessiano y puntos críticos

Un punto crítico satisface:

$$
\boxed{
\nabla f(x,y)=0
}
$$

Por ejemplo:

$$
f(x,y)=x^2+3y^2.
$$

Tenemos:

$$
\nabla f=
\begin{bmatrix}
2x\\
6y
\end{bmatrix}.
$$

Para encontrar el punto crítico:

$$
2x=0
$$

$$
6y=0
$$

por lo tanto:

$$
\boxed{(x,y)=(0,0)}
$$

Ahora necesitamos saber qué tipo de punto es.

Aquí entra el Hessiano.

# 7. Clasificación mediante el Hessiano

Para:

$$
H=
\begin{bmatrix}
f_{xx}&f_{xy}\\
f_{yx}&f_{yy}
\end{bmatrix}
$$

calculamos:

$$
D=
\det(H)
$$

es decir:

$$
D=f_{xx}f_{yy}-f_{xy}^2.
$$

Entonces:

### Caso 1

Si:

$$
D>0
$$

y:

$$
f_{xx}>0
$$

tenemos:

$$
\boxed{\text{mínimo local}}
$$

### Caso 2

Si:

$$
D>0
$$

y:

$$
f_{xx}<0
$$

tenemos:

$$
\boxed{\text{máximo local}}
$$

### Caso 3

Si:

$$
D<0
$$

tenemos:

$$
\boxed{\text{punto silla}}
$$

### Caso 4

Si:

$$
D=0
$$

el criterio no decide.

Hay que utilizar otro análisis.

# 8. Ejemplo difícil: punto silla

Consideremos:

$$
f(x,y)=x^2-y^2.
$$

Tenemos:

$$
f_x=2x
$$

$$
f_y=-2y.
$$

El único punto crítico es:

$$
(0,0).
$$

El Hessiano es:

$$
\boxed{
H=
\begin{bmatrix}
2&0\\
0&-2
\end{bmatrix}
}
$$

El determinante:

$$
D=(2)(-2)-0
$$

$$
D=-4.
$$

Como:

$$
D<0
$$

tenemos:

$$
\boxed{\text{punto silla}}
$$

# 9. ¿Por qué se llama punto silla?

Porque en una dirección la superficie sube y en otra baja.

Si:

$$
y=0
$$

tenemos:

$$
f(x,0)=x^2
$$

que es una parábola hacia arriba.

Pero si:

$$
x=0
$$

tenemos:

$$
f(0,y)=-y^2
$$

que es una parábola hacia abajo.

Por eso:

$$
\boxed{\text{sube en una dirección y baja en otra}}
$$

# 11. Segundo orden: la idea fundamental

Ahora llegamos a una de las partes más importantes.

En una variable conoces la expansión de Taylor:

$$
f(x+\Delta x)
\approx
f(x)
+
f'(x)\Delta x
+
\frac12f''(x)(\Delta x)^2.
$$

En varias variables ocurre algo parecido.

La expansión de segundo orden es:

$$
\boxed{
f(x+\Delta x)
\approx
f(x)
+
\nabla f(x)^T\Delta x
+
\frac12
\Delta x^T
H(x)
\Delta x
}
$$

Esta fórmula debes aprenderla muy bien.

# 12. ¿Qué representa cada término?

Tenemos:

$$
f(x)
$$

= valor actual.

Luego:

$$
\nabla f^T\Delta x
$$

= cambio lineal.

Y:

$$
\frac12\Delta x^TH\Delta x
$$

= corrección por curvatura.

Por eso:

$$
\boxed{
\text{Taylor de segundo orden}
=
\text{valor}
+
\text{pendiente}
+
\text{curvatura}
}
$$

# 13. Ejemplo de segundo orden

Consideremos:

$$
f(x,y)
=
x^2+4xy+5y^2.
$$

El gradiente es:

$$
\nabla f=
\begin{bmatrix}
2x+4y\\
4x+10y
\end{bmatrix}.
$$

El Hessiano:

$$
\boxed{
H=
\begin{bmatrix}
2&4\\
4&10
\end{bmatrix}
}
$$

Observa que en este caso el Hessiano es constante.

# 14. Evaluemos en un punto

Tomemos:

$$
x_0=
\begin{bmatrix}
1\\
2
\end{bmatrix}.
$$

Tenemos:

$$
f(1,2)
=
1+8+20
=
29.
$$

El gradiente:

$$
\nabla f(1,2)
=
\begin{bmatrix}
2+8\\
4+20
\end{bmatrix}
=
\begin{bmatrix}
10\\
24
\end{bmatrix}.
$$

Y:

$$
H=
\begin{bmatrix}
2&4\\
4&10
\end{bmatrix}.
$$

Ahora supongamos:

$$
\Delta x=
\begin{bmatrix}
0.1\\
-0.05
\end{bmatrix}.
$$

La aproximación de segundo orden es:

$$
f(x_0+\Delta x)
\approx
29
+
\begin{bmatrix}
10&24
\end{bmatrix}
\begin{bmatrix}
0.1\\
-0.05
\end{bmatrix}
+
\frac12
\begin{bmatrix}
0.1&-0.05
\end{bmatrix}
H
\begin{bmatrix}
0.1\\
-0.05
\end{bmatrix}.
$$

Como $(f)$ es cuadrática, esta expansión es **exacta**.

Esta observación es muy importante:

$$
\boxed{
\text{para funciones cuadráticas, Taylor de segundo orden es exacto}
}
$$

# 15. ¿Por qué el Hessiano es importante para optimización?

Supongamos que queremos minimizar:

$$
f(x).
$$

Gradient Descent utiliza:

$$
x_{k+1}
=
x_k-\eta\nabla f(x_k).
$$

Esto utiliza solamente información de **primer orden**.

Newton utiliza además:

$$
H(x).
$$

Es decir, conoce la curvatura.

# 16. Newton en una variable

Recuerda primero:

$$
f'(x)=0
$$

es la condición de mínimo/máximo.

Newton aproxima:

$$
f(x+\Delta x)
$$

mediante:

$$
f(x)
+
f'(x)\Delta x
+
\frac12f''(x)(\Delta x)^2.
$$

Buscando el mínimo de esta aproximación:

$$
f'(x)+f''(x)\Delta x=0.
$$

Por tanto:

$$
\boxed{
\Delta x
=
-\frac{f'(x)}{f''(x)}
}
$$

y:

$$
\boxed{
x_{k+1}
=
x_k-\frac{f'(x_k)}{f''(x_k)}
}
$$

# 17. Newton multivariable

En varias variables:

$$
\nabla f(x)=0.
$$

Usando Taylor:

$$
\nabla f(x+\Delta x)
\approx
\nabla f(x)
+
H(x)\Delta x.
$$

Queremos:

$$
\nabla f(x+\Delta x)=0.
$$

Entonces:

$$
\nabla f(x)+H(x)\Delta x=0.
$$

Por tanto:

$$
\boxed{
H(x)\Delta x
=
-\nabla f(x)
}
$$

y:

$$
\boxed{
x_{k+1}=x_k+\Delta x
}
$$

Esta es la forma numéricamente correcta de pensar Newton.

No necesitamos calcular:

$$
H^{-1}.
$$

Resolvemos un sistema lineal.

# 18. Ejemplo difícil de Newton

Consideremos:

$$
f(x,y)
=
x^4+y^4-3x^2-2y^2+xy.
$$

Primero:

$$
f_x=
4x^3-6x+y
$$

$$
f_y=
4y^3-4y+x.
$$

Por tanto:

$$
\nabla f=
\begin{bmatrix}
4x^3-6x+y\\
4y^3-4y+x
\end{bmatrix}.
$$

El Hessiano es:

$$
\boxed{
H=
\begin{bmatrix}
12x^2-6&1\\
1&12y^2-4
\end{bmatrix}
}
$$

Newton realiza:

$$
\begin{bmatrix}
12x^2-6&1\\
1&12y^2-4
\end{bmatrix}
\begin{bmatrix}
\Delta x\\
\Delta y
\end{bmatrix}
=
-
\begin{bmatrix}
4x^3-6x+y\\
4y^3-4y+x
\end{bmatrix}.
$$

# 19. Implementación de Newton en Python

````python
import numpy as np

def gradient(x):
    x1, x2 = x

    return np.array([
        4*x1**3 - 6*x1 + x2,
        4*x2**3 - 4*x2 + x1
    ])


def hessian(x):
    x1, x2 = x

    return np.array([
        [12*x1**2 - 6, 1],
        [1, 12*x2**2 - 4]
    ])


x = np.array([2.0, 2.0])

trajectory = [x.copy()]

for k in range(20):

    g = gradient(x)
    H = hessian(x)

    # H Δx = -g
    delta = np.linalg.solve(H, -g)

    x = x + delta

    trajectory.append(x.copy())

    if np.linalg.norm(g) < 1e-10:
        break

trajectory = np.array(trajectory)

print("Solución:")
print(x)

print("Gradiente final:")
print(gradient(x))
````

# 20. Newton vs Gradient Descent

Esta comparación es fundamental para Machine Learning.

### Gradient Descent

Utiliza:

$$
\nabla f
$$

y:

$$
x_{k+1}=x_k-\eta\nabla f.
$$

No conoce directamente la curvatura.

### Newton

Utiliza:

$$
\nabla f
$$

y:

$$
H.
$$

Resuelve:

$$
H\Delta x=-\nabla f.
$$

Por tanto:

$$
\boxed{
\text{Newton conoce la geometría local de la función}
}
$$

# 23. Eigenvalores del Hessiano

Aquí se conecta con álgebra lineal.

Si:

$$
H
$$

es simétrico, podemos obtener:

$$
H=Q\Lambda Q^T.
$$

Los eigenvalores:

$$
\lambda_1,\lambda_2,\ldots,\lambda_n
$$

representan la curvatura en determinadas direcciones principales.

Por ejemplo:

$$
H=
\begin{bmatrix}
2&0\\
0&200
\end{bmatrix}.
$$

Tiene eigenvalores:

$$
\boxed{2,\quad200}
$$

Por eso sabemos que la función tiene una curvatura mucho mayor en una dirección.

# 24. Hessiano positivo definido

Este concepto es absolutamente fundamental.

Decimos:

$$
\boxed{H\succ0}
$$

si:

$$
x^THx>0
$$

para todo:

$$
x\neq0.
$$

Si el Hessiano es positivo definido cerca de un punto crítico:

$$
\boxed{\text{tenemos un mínimo local}}
$$

Intuitivamente:

$$
\boxed{
\text{la superficie tiene forma de cuenco}
}
$$

# 25. Hessiano negativo definido

Si:

$$
x^THx<0
$$

para todo:

$$
x\neq0
$$

entonces:

$$
\boxed{H\prec0}
$$

y tenemos una superficie tipo cúpula.

Por tanto:

$$
\boxed{\text{máximo local}}
$$

# 26. Hessiano indefinido

Si existen direcciones donde:

$$
x^THx>0
$$

y otras donde:

$$
x^THx<0,
$$

entonces:

$$
\boxed{\text{Hessian indefinido}}
$$

y tenemos un punto silla.

# 27. Tabla fundamental

| Hessiano en punto crítico | Resultado           |
| ------------------------- | ------------------- |
| $(H\succ0)$               | mínimo local        |
| $(H\prec0)$               | máximo local        |
| $(H)$ indefinido          | punto silla         |
| $(H)$ semidefinido        | criterio inconcluso |

En términos de eigenvalores:

| Eigenvalores          | Interpretación       |
| --------------------- | -------------------- |
| todos $(>0)$          | mínimo               |
| todos $(<0)$          | máximo               |
| algunos + y algunos − | silla                |
| alguno = 0            | puede ser inconcluso |

Imagina una fábrica que produce un producto.

$$

$$
C(x,y)
=

te indica cómo cambiar los parámetros.

elementos.

# 30. Análisis de estabilidad

Ahora llegamos a la última parte que pediste.

Consideremos una ecuación diferencial:

$$
\boxed{
\frac{dx}{dt}=F(x)
}
$$

Supongamos que:

$$
x^*
$$

es un equilibrio:

$$
F(x^*)=0.
$$

Queremos saber:

> ¿Qué ocurre si perturbamos ligeramente el sistema?

Escribimos:

$$
x=x^*+\delta x.
$$

Taylor de primer orden:

$$
F(x^*+\delta x)
\approx
F(x^*)+
J_F(x^*)\delta x.
$$

Como:

$$
F(x^*)=0,
$$

obtenemos:

$$
\boxed{
\frac{d\delta x}{dt}
=
J_F(x^*)\delta x
}
$$

Ahora tenemos un sistema lineal.

# 31. ¿Qué determina la estabilidad?

Los eigenvalores:

$$
\lambda_i
$$

del Jacobiano.

Si:

$$
\boxed{
\operatorname{Re}(\lambda_i)<0
}
$$

para todos:

$$
\boxed{\text{estable asintóticamente}}
$$

Si alguno tiene:

$$
\boxed{
\operatorname{Re}(\lambda_i)>0
}
$$

el equilibrio es inestable.

# 32. Ejemplo sencillo

Consideremos:

$$
\frac{dx}{dt}=-2x.
$$

El equilibrio es:

$$
x^*=0.
$$

Tenemos:

$$
F'(x)=-2.
$$

Por tanto:

$$
\lambda=-2.
$$

Como:

$$
\lambda<0
$$

el sistema es estable.

La solución:

$$
x(t)=x_0e^{-2t}.
$$

Y:

$$
e^{-2t}\rightarrow0.
$$

Por eso cualquier perturbación desaparece.

# 33. Sistema inestable

Ahora:

$$
\frac{dx}{dt}=2x.
$$

Tenemos:

$$
\lambda=2>0.
$$

La solución:

$$
x(t)=x_0e^{2t}.
$$

La perturbación crece.

Por tanto:

$$
\boxed{\text{inestable}}
$$

# 34. Python: comparar estabilidad

````python
import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(0, 5, 500)

x0 = 0.1

x_stable = x0 * np.exp(-2*t)
x_unstable = x0 * np.exp(2*t)

plt.figure(figsize=(10, 6))

plt.plot(
    t,
    x_stable,
    label="Estable: λ = -2"
)

plt.plot(
    t,
    x_unstable,
    label="Inestable: λ = 2"
)

plt.axhline(0)

plt.xlabel("Tiempo")
plt.ylabel("Perturbación")

plt.title("Estabilidad determinada por el eigenvalor")

plt.legend()
plt.grid()

plt.show()
````

# 35. Conexión profunda: Hessiano y estabilidad

Hay dos situaciones parecidas pero conceptualmente diferentes.

## Optimización

Tenemos:

$$
f(x)
$$

y estudiamos:

$$
H=\nabla^2f.
$$

Queremos determinar si un punto es:

$$
\text{mínimo, máximo o silla}.
$$

## Sistemas dinámicos

Tenemos:

$$
\dot{x}=F(x)
$$

y estudiamos:

$$
J_F.
$$

Queremos determinar si un equilibrio es:

$$
\text{estable o inestable}.
$$

Pero hay una conexión importante.

Si el sistema es un **gradient flow**:

$$
\boxed{
\dot{x}=-\nabla f(x)
}
$$

entonces:

$$
J_F=-H_f.
$$

¡Aquí aparece directamente el Hessiano!

# 36. Gradient flow

Consideremos:

$$
\dot{x}=-\nabla f(x).
$$

La energía es:

$$
f(x).
$$

Derivamos respecto al tiempo:

$$
\frac{df}{dt}
=
\nabla f(x)^T\dot{x}.
$$

Como:

$$
\dot{x}=-\nabla f
$$

tenemos:

$$
\frac{df}{dt}
=
-\nabla f^T\nabla f
$$

por tanto:

$$
\boxed{
\frac{df}{dt}
=
-\|\nabla f\|^2\le0
}
$$

Esto es una ecuación increíblemente importante.

Significa que el sistema siempre disminuye la función:

$$
f.
$$

Por eso gradient descent puede verse como una discretización de una dinámica continua.

# 37. Hessiano + Gradient Flow

Supongamos un punto crítico:

$$
\nabla f(x^*)=0.
$$

El sistema:

$$
\dot{x}=-\nabla f(x)
$$

tiene equilibrio en $(x^*)$.

El Jacobiano del sistema es:

$$
J_F=-H_f.
$$

Si:

$$
H_f\succ0
$$

entonces sus eigenvalores son:

$$
\lambda_i(H)>0.
$$

Pero los eigenvalores de:

$$
-H
$$

son:

$$
-\lambda_i.
$$

Por tanto:

$$
-\lambda_i<0.
$$

Entonces:

$$
\boxed{
\text{mínimo de }f
\Longleftrightarrow
\text{equilibrio estable del gradient flow}
}
$$

Esta conexión es muy importante para SciML.

# 38. Un ejemplo completo de SciML

Supongamos:

$$
f(x,y)
=
x^2+5y^2.
$$

Tenemos:

$$
\nabla f=
\begin{bmatrix}
2x\\
10y
\end{bmatrix}
$$

y:

$$
H=
\begin{bmatrix}
2&0\\
0&10
\end{bmatrix}.
$$

El gradient flow es:

$$
\frac{dx}{dt}=-2x
$$

$$
\frac{dy}{dt}=-10y.
$$

Las soluciones son:

$$
x(t)=x_0e^{-2t}
$$

$$
y(t)=y_0e^{-10t}.
$$

Ambas tienden a:

$$
(0,0).
$$

Por tanto:

$$
\boxed{(0,0)\text{ es estable}}
$$

y es precisamente el mínimo de:

$$
f(x,y).
$$

# 39. Visualización Python del gradient flow

````python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-4, 4, 300)
y = np.linspace(-4, 4, 300)

X, Y = np.meshgrid(x, y)

Z = X**2 + 5*Y**2

# Campo gradient flow: dx/dt = -grad f
U = -2*X
V = -10*Y

plt.figure(figsize=(10, 8))

contours = plt.contour(
    X, Y, Z,
    levels=20
)

plt.clabel(contours, inline=True, fontsize=8)

plt.streamplot(
    X, Y,
    U, V,
    density=1.5
)

plt.scatter(
    [0],
    [0],
    s=150
)

plt.xlabel("x")
plt.ylabel("y")

plt.title("Gradient Flow y estabilidad")

plt.axis("equal")
plt.grid()

plt.show()
````

Las trayectorias fluyen hacia el mínimo.
