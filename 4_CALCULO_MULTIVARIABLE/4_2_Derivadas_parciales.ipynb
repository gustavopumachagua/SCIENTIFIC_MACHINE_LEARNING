# 1. La idea central

Supongamos que tenemos una función de dos variables:

$$
f(x,y)
$$

Geométricamente, esto representa una **superficie**:

$$
z=f(x,y)
$$

Por ejemplo:

$$
f(x,y)=x^2+2y^2
$$

Si estás en un punto $(x_0,y_0)$, puedes preguntarte tres cosas diferentes:

### Pregunta 1 — ¿Qué pasa si solamente cambio $(x)$?

Eso es una **derivada parcial**:

$$
\boxed{\frac{\partial f}{\partial x}}
$$

### Pregunta 2 — ¿Qué pasa si me muevo en una dirección arbitraria?

Eso es una **derivada direccional**:

$$
\boxed{D_{\mathbf u}f}
$$

### Pregunta 3 — ¿Cuál es la dirección en la que $(f)$ aumenta más rápidamente?

Eso lo responde el **gradiente**:

$$
\boxed{\nabla f}
$$

Y existe una relación fundamental:

$$
\boxed{D_{\mathbf u}f=\nabla f\cdot\mathbf u}
$$

Esta ecuación es una de las más importantes de todo el cálculo multivariable.

# 2. Derivadas parciales

## 2.1 ¿Qué es una derivada parcial?

Consideremos:

$$
f(x,y)=x^2+3xy+2y^2
$$

Una derivada parcial pregunta:

> ¿Cómo cambia $(f)$ respecto a $(x)$, manteniendo $(y)$ constante?

Escribimos:

$$
\frac{\partial f}{\partial x}
$$

Cuando derivamos respecto a $(x)$, tratamos $(y)$ como una constante.

Entonces:

$$
\frac{\partial}{\partial x}(x^2)=2x
$$

$$
\frac{\partial}{\partial x}(3xy)=3y
$$

porque $(y)$ se considera constante.

Y:

$$
\frac{\partial}{\partial x}(2y^2)=0
$$

Por tanto:

$$
\boxed{f_x=2x+3y}
$$

Ahora respecto a $(y)$:

$$
\frac{\partial}{\partial y}(x^2)=0
$$

$$
\frac{\partial}{\partial y}(3xy)=3x
$$

$$
\frac{\partial}{\partial y}(2y^2)=4y
$$

por lo que:

$$
\boxed{f_y=3x+4y}
$$

# 3. Interpretación geométrica

Supongamos:

$$
z=f(x,y)
$$

La superficie tiene dos coordenadas horizontales:

$$
x,\quad y
$$

y una altura:

$$
z.
$$

La derivada parcial $(f_x)$ mide la pendiente si te desplazas **solamente en dirección $(x)$**.

La derivada parcial $(f_y)$ mide la pendiente si te desplazas **solamente en dirección $(y)$**.

Es parecido a caminar sobre una montaña.

Si caminas hacia el este:

$$
\frac{\partial f}{\partial x}
$$

te dice qué tan inclinada está la montaña en esa dirección.

Si caminas hacia el norte:

$$
\frac{\partial f}{\partial y}
$$

te dice qué tan inclinada está en esa dirección.

# 4. Ejemplo difícil: temperatura en una habitación

Supongamos que la temperatura de una habitación está modelada por:

$$
T(x,y)
=
20
+4x
-3y
-0.5x^2
-0.8y^2
+0.6xy
$$

donde:

* $(x)$ = posición horizontal en metros
* $(y)$ = posición vertical en metros
* $(T)$ = temperatura en °C.

Queremos estudiar qué ocurre en:

$$
(x,y)=(2,3)
$$

## 4.1 Derivada respecto a $(x)$

Tenemos:

$$
T_x=
4-x+0.6y
$$

porque:

$$
\frac{\partial}{\partial x}(-0.5x^2)=-x
$$

y:

$$
\frac{\partial}{\partial x}(0.6xy)=0.6y
$$

Ahora evaluamos:

$$
T_x(2,3)
=
4-2+0.6(3)
$$

$$
=2+1.8
$$

$$
\boxed{T_x(2,3)=3.8}
$$

Interpretación:

> Si en $(2,3)$ avanzamos ligeramente en dirección $(x)$, la temperatura aumenta aproximadamente $(3.8,^\circ C)$ por metro.

## 4.2 Derivada respecto a $(y)$

Tenemos:

$$
T_y=-3-1.6y+0.6x
$$

Evaluando:

$$
T_y(2,3)
=
-3-1.6(3)+0.6(2)
$$

$$
=-3-4.8+1.2
$$

$$
\boxed{T_y(2,3)=-6.6}
$$

Interpretación:

> Si avanzamos en dirección $(y)$, la temperatura disminuye aproximadamente $(6.6,^\circ C)$ por metro.

Esto ya es interesante.

En el punto $(2,3)$:

$$
T_x=3.8
$$

pero:

$$
T_y=-6.6.
$$

La temperatura está aumentando en $(x)$, pero disminuyendo en $(y)$.

La gráfica representa:

$$
z=T(x,y)
$$

# 6. ¿Por qué necesitamos derivadas direccionales?

Las derivadas parciales solamente estudian dos direcciones particulares:

$$
(1,0)
$$

y:

$$
(0,1)
$$

Pero en la vida real normalmente no nos movemos exactamente en esas direcciones.

Supongamos que estás caminando diagonalmente.

Tu dirección podría ser:

$$
\mathbf v=(3,4)
$$

Entonces queremos saber:

> ¿Qué tan rápido cambia la temperatura mientras camino en esa dirección?

Ahí aparecen las **derivadas direccionales**.

# 7. Derivada direccional

Sea:

$$
f(x,y)
$$

y sea:

$$
\mathbf u=(u_1,u_2)
$$

un vector unitario.

La derivada direccional es:

$$
\boxed{
D_{\mathbf u}f
=
\nabla f\cdot\mathbf u
}
$$

donde:

$$
\nabla f=
\begin{bmatrix}
f_x\\
f_y
\end{bmatrix}.
$$

Es decir:

$$
D_{\mathbf u}f
=
f_xu_1+f_yu_2
$$

# 8. ¡Muy importante!: la dirección debe ser unitaria

Supongamos:

$$
\mathbf v=(3,4)
$$

Su norma es:

$$
\|\mathbf v\|
=
\sqrt{3^2+4^2}
=
5
$$

Entonces el vector unitario es:

$$
\mathbf u=
\frac{\mathbf v}{\|\mathbf v\|}
$$

$$
\boxed{
\mathbf u=
\left(
\frac35,\frac45
\right)
}
$$

Esto es:

$$
\mathbf u=(0.6,0.8)
$$

# 9. Ejemplo difícil de derivada direccional

Tomemos:

$$
f(x,y)=x^3+2xy^2-4x^2y+5y^3
$$

Queremos calcular la derivada direccional en:

$$
(1,2)
$$

en la dirección:

$$
\mathbf v=(3,-4)
$$

## Paso 1: derivadas parciales

Respecto a $(x)$:

$$
f_x
=
3x^2+2y^2-8xy
$$

Respecto a $(y)$:

$$
f_y
=
4xy-4x^2+15y^2
$$

## Paso 2: evaluar en $(1,2)$

Para $(f_x)$:

$$
f_x(1,2)
=
3(1)^2+2(2)^2-8(1)(2)
$$

$$
=3+8-16
$$

$$
\boxed{f_x(1,2)=-5}
$$

Ahora:

$$
f_y(1,2)
=
4(1)(2)-4(1)^2+15(2)^2
$$

$$
=8-4+60
$$

$$
\boxed{f_y(1,2)=64}
$$

Por tanto:

$$
\boxed{
\nabla f(1,2)=(-5,64)
}
$$

# 10. Normalizamos la dirección

Tenemos:

$$
\mathbf v=(3,-4)
$$

Su norma:

$$
\|\mathbf v\|=5
$$

por tanto:

$$
\mathbf u=
\left(
\frac35,-\frac45
\right)
$$

# 11. Calculamos la derivada direccional

$$
D_{\mathbf u}f
=
(-5,64)
\cdot
\left(
\frac35,-\frac45
\right)
$$

Entonces:

$$
D_{\mathbf u}f
=
-5\frac35
+
64\left(-\frac45\right)
$$

$$
=-3-51.2
$$

$$
\boxed{
D_{\mathbf u}f=-54.2
}
$$

Interpretación:

> Si nos movemos desde $(1,2)$ en la dirección $(3,-4)$, la función disminuye aproximadamente $(54.2)$ unidades por unidad de distancia.

Esto es mucho más poderoso que conocer solamente $(f_x)$ y $(f_y)$.

# 12. El gradiente

Ahora llegamos al concepto central.

El gradiente es:

$$
\boxed{
\nabla f=
\begin{bmatrix}
\frac{\partial f}{\partial x}\\
\frac{\partial f}{\partial y}
\end{bmatrix}
}
$$

Para:

$$
f(x,y)=x^3+2xy^2-4x^2y+5y^3
$$

tenemos:

$$
\boxed{
\nabla f(x,y)
=
\begin{bmatrix}
3x^2+2y^2-8xy\\
4xy-4x^2+15y^2
\end{bmatrix}
}
$$

# 13. ¿Qué significa físicamente el gradiente?

Esta es una idea extremadamente importante:

$$
\boxed{\nabla f}
$$

apunta hacia la dirección donde $(f)$ **crece más rápidamente**.

Y:

$$
\boxed{\|\nabla f\|}
$$

es la máxima tasa de crecimiento posible.

Por ejemplo, si:

$$
\nabla f=(3,4)
$$

entonces:

$$
\|\nabla f\|
=
\sqrt{3^2+4^2}
=
5
$$

Eso significa:

> Existe una dirección en la que $(f)$ aumenta a una velocidad de 5 unidades por unidad de distancia, y esa dirección es la dirección del gradiente.

La dirección de máximo crecimiento es:

$$
\boxed{
\frac{\nabla f}{\|\nabla f\|}
}
$$

# 14. Relación fundamental

La derivada direccional es:

$$
D_{\mathbf u}f
=
\nabla f\cdot\mathbf u
$$

Usando la definición del producto punto:

$$
\nabla f\cdot\mathbf u
=
\|\nabla f\|
\|\mathbf u\|
\cos\theta
$$

Como $(\mathbf u)$ es unitario:

$$
\|\mathbf u\|=1
$$

entonces:

$$
\boxed{
D_{\mathbf u}f
=
\|\nabla f\|\cos\theta
}
$$

Esta fórmula explica prácticamente todo.

# 15. ¿Cuándo tenemos máximo crecimiento?

Si:

$$
\theta=0
$$

entonces:

$$
\cos(0)=1
$$

por tanto:

$$
D_{\mathbf u}f
=
\|\nabla f\|
$$

Así:

$$
\boxed{
D_{\mathbf u}f_{\max}=\|\nabla f\|
}
$$

cuando:

$$
\boxed{
\mathbf u=
\frac{\nabla f}{\|\nabla f\|}
}
$$

# 16. ¿Cuándo tenemos máximo decrecimiento?

Si:

$$
\theta=\pi
$$

entonces:

$$
\cos(\pi)=-1
$$

y:

$$
\boxed{
D_{\mathbf u}f_{\min}
=
-\|\nabla f\|
}
$$

La dirección es:

$$
\boxed{
-\frac{\nabla f}{\|\nabla f\|}
}
$$

# 17. ¿Cuándo no cambia la función?

Si:

$$
\theta=\frac{\pi}{2}
$$

entonces:

$$
\cos\left(\frac{\pi}{2}\right)=0
$$

por tanto:

$$
\boxed{
D_{\mathbf u}f=0
}
$$

Esto significa que te estás desplazando en una dirección donde, localmente, la función no cambia.

Esto nos lleva a las **curvas de nivel**.

# 18. Gradiente y curvas de nivel

Una curva de nivel satisface:

$$
f(x,y)=c
$$

Por ejemplo:

$$
f(x,y)=x^2+y^2
$$

Las curvas de nivel son:

$$
x^2+y^2=c
$$

que son círculos.

El gradiente es:

$$
\nabla f=(2x,2y)
$$

Observa que el vector:

$$
(2x,2y)
$$

apunta radialmente hacia afuera.

Por tanto es perpendicular a los círculos.

La propiedad general es:

$$
\boxed{
\nabla f\perp\text{curva de nivel}
}
$$

Esta propiedad aparece constantemente en física.

# 19. Gráfica: curvas de nivel + gradiente

````python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-4, 4, 300)
y = np.linspace(-4, 4, 300)

X, Y = np.meshgrid(x, y)

# Función
F = X**2 + Y**2

# Gradiente
Fx = 2*X
Fy = 2*Y

plt.figure(figsize=(9, 7))

# Curvas de nivel
contours = plt.contour(
    X, Y, F,
    levels=15
)

plt.clabel(contours, inline=True, fontsize=8)

# Vectores del gradiente
skip = 20

plt.quiver(
    X[::skip, ::skip],
    Y[::skip, ::skip],
    Fx[::skip, ::skip],
    Fy[::skip, ::skip]
)

plt.xlabel("x")
plt.ylabel("y")
plt.title("Gradiente y curvas de nivel")

plt.axis("equal")
plt.grid()

plt.show()
````

Observa conceptualmente que los vectores del gradiente atraviesan las curvas de nivel perpendicularmente.

# 21. Ejemplo científico: temperatura

Supongamos:

$$
T(x,y,z)
$$

representa la temperatura dentro de un material.

Entonces:

$$
\nabla T=
\begin{bmatrix}
T_x\\
T_y\\
T_z
\end{bmatrix}
$$

indica:

> dirección de máximo aumento de temperatura.

Y aquí aparece una conexión muy importante con física.

La ley de Fourier establece:

$$
\boxed{
\mathbf q=-k\nabla T
}
$$

donde:

* $(\mathbf q)$ = flujo de calor;
* $(k)$ = conductividad térmica;
* $(\nabla T)$ = gradiente de temperatura.

El signo negativo significa que:

$$
\boxed{\text{el calor fluye hacia temperaturas menores}}
$$

Esto conecta directamente:

$$
\boxed{
\text{derivadas parciales}
\rightarrow
\text{gradiente}
\rightarrow
\text{física}
\rightarrow
\text{PDE}
\rightarrow
\text{SciML}
}
$$

# 22. Ejemplo de SciML: ecuación del calor

La ecuación del calor es:

$$
\boxed{
\frac{\partial T}{\partial t}
=
\alpha\nabla^2T
}
$$

En 2D:

$$
\frac{\partial T}{\partial t}
=
\alpha
\left(
\frac{\partial^2T}{\partial x^2}
+
\frac{\partial^2T}{\partial y^2}
\right)
$$

Observa que para llegar a esta ecuación necesitamos:

### Primera capa

Derivadas parciales:

$$
T_x,\quad T_y
$$

### Segunda capa

Segundas derivadas:

$$
T_{xx},\quad T_{yy}
$$

### Tercera capa

Laplaciano:

$$
\nabla^2T=T_{xx}+T_{yy}
$$

### Cuarta capa

PDE:

$$
T_t=\alpha\nabla^2T
$$

Esto es exactamente el tipo de matemática que posteriormente utilizarás en SciML.

# 24. Python: visualizar temperatura y gradiente

````python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-4, 4, 250)
y = np.linspace(-4, 4, 250)

X, Y = np.meshgrid(x, y)

# Temperatura
T = 100 - 2*X**2 - 3*Y**2 + 4*X*Y

# Derivadas parciales
Tx = -4*X + 4*Y
Ty = -6*Y + 4*X

plt.figure(figsize=(10, 8))

# Mapa de temperatura
plt.contourf(X, Y, T, levels=30)

# Curvas de nivel
contours = plt.contour(
    X, Y, T,
    levels=15
)

plt.clabel(contours, inline=True, fontsize=8)

# Gradiente
skip = 15

plt.quiver(
    X[::skip, ::skip],
    Y[::skip, ::skip],
    Tx[::skip, ::skip],
    Ty[::skip, ::skip]
)

# Punto de análisis
plt.scatter(
    [2], [1],
    s=100
)

plt.xlabel("x")
plt.ylabel("y")
plt.title("Temperatura y campo del gradiente")

plt.axis("equal")
plt.grid()

plt.show()
````

# 25. Gradiente en Machine Learning

Aquí es donde este tema se vuelve **fundamental para SciML**.

Supongamos una función de pérdida:

$$
L(w_1,w_2)
$$

Por ejemplo:

$$
L(w_1,w_2)
=
(w_1-3)^2
+
5(w_2+2)^2
$$

Tenemos:

$$
\frac{\partial L}{\partial w_1}
=
2(w_1-3)
$$

$$
\frac{\partial L}{\partial w_2}
=
10(w_2+2)
$$

Por tanto:

$$
\boxed{
\nabla L=
\begin{bmatrix}
2(w_1-3)\\
10(w_2+2)
\end{bmatrix}
}
$$

# 26. Gradient Descent

La idea es:

$$
\boxed{
\mathbf w_{k+1}
=
\mathbf w_k-\eta\nabla L(\mathbf w_k)
}
$$

¿Por qué usamos el signo negativo?

Porque el gradiente apunta hacia donde:

$$
L
$$

aumenta más rápidamente.

Entonces:

$$
-\nabla L
$$

apunta hacia donde disminuye más rápidamente.

Esto es la base del:

$$
\boxed{\text{Gradient Descent}}
$$

que posteriormente aparece en:

* regresión;
* redes neuronales;
* PINNs;
* Neural ODEs;
* DeepONet;
* operadores neuronales;
* calibración de modelos físicos.

# 27. Visualización de Gradient Descent en Python

````python
import numpy as np
import matplotlib.pyplot as plt

def loss(x, y):
    return (x - 3)**2 + 5*(y + 2)**2

def gradient(x, y):
    return np.array([
        2*(x - 3),
        10*(y + 2)
    ])

# Malla
x = np.linspace(-4, 7, 300)
y = np.linspace(-6, 3, 300)

X, Y = np.meshgrid(x, y)

Z = loss(X, Y)

# Parámetros iniciales
w = np.array([-3.0, 2.5])

learning_rate = 0.08

trajectory = [w.copy()]

for _ in range(40):

    grad = gradient(w[0], w[1])

    w = w - learning_rate * grad

    trajectory.append(w.copy())

trajectory = np.array(trajectory)

# Gráfica
plt.figure(figsize=(10, 8))

contours = plt.contour(
    X, Y, Z,
    levels=20
)

plt.clabel(contours, inline=True, fontsize=8)

plt.plot(
    trajectory[:, 0],
    trajectory[:, 1],
    marker="o"
)

plt.scatter(
    [3],
    [-2],
    s=150
)

plt.xlabel("$w_1$")
plt.ylabel("$w_2$")

plt.title("Trayectoria del Gradient Descent")

plt.grid()

plt.show()
````

Observa cómo el algoritmo intenta llegar al mínimo:

$$
\boxed{(3,-2)}
$$

# 28. Una conexión todavía más profunda con SciML

Supongamos que tienes una ecuación física:

$$
\frac{du}{dt}=F(u,\theta)
$$

donde $(\theta)$ es un parámetro desconocido.

Por ejemplo:

$$
\frac{du}{dt}=-ku
$$

y queremos encontrar:

$$
k.
$$

Tenemos datos experimentales:

$$
D=\{t_i,y_i\}_{i=1}^{N}
$$

y definimos una función de pérdida:

$$
L(k)
=
\sum_{i=1}^{N}
\left[
u(t_i;k)-y_i
\right]^2
$$

Ahora necesitamos:

$$
\frac{dL}{dk}
$$

para encontrar el valor de $(k)$ que mejor reproduce los experimentos.

En problemas reales podrías tener:

$$
\theta=
(\alpha,k,E,\rho,c_p,\ldots)
$$

y entonces:

$$
L=L(\theta_1,\theta_2,\ldots,\theta_n)
$$

El gradiente es:

$$
\boxed{
\nabla_\theta L
=
\begin{bmatrix}
\frac{\partial L}{\partial\theta_1}\\
\frac{\partial L}{\partial\theta_2}\\
\vdots\\
\frac{\partial L}{\partial\theta_n}
\end{bmatrix}
}
$$

y puedes optimizar:

$$
\boxed{
\theta_{k+1}
=
\theta_k-\eta\nabla_\theta L
}
$$

Esto es **parameter estimation** dentro de SciML.

# 29. PINNs: aquí las derivadas parciales son todavía más importantes

Supongamos que queremos resolver:

$$
u_t+\alpha u_{xx}=0
$$

utilizando una red neuronal:

$$
u_\theta(x,t)
$$

La red recibe:

$$
(x,t)
$$

y produce:

$$
u_\theta(x,t).
$$

Pero necesitamos calcular:

$$
\frac{\partial u_\theta}{\partial t}
$$

y:

$$
\frac{\partial^2u_\theta}{\partial x^2}.
$$

Construimos el residuo:

$$
\boxed{
r_\theta(x,t)
=
u_t+\alpha u_{xx}
}
$$

y queremos:

$$
r_\theta\approx0.
$$

Por eso una PINN necesita cálculo multivariable y diferenciación automática.

Aquí la cadena conceptual es:

$$
\boxed{
\text{derivadas parciales}
\rightarrow
\text{gradiente}
\rightarrow
\text{derivadas de redes}
\rightarrow
\text{residuo PDE}
\rightarrow
\text{PINNs}
}
$$

# 30. Diferencia fundamental entre los tres conceptos

| Concepto                  | Pregunta                                             |
| ------------------------- | ---------------------------------------------------- |
| $(\partial f/\partial x)$ | ¿Qué pasa si cambio $(x)$ solamente?                 |
| $(\partial f/\partial y)$ | ¿Qué pasa si cambio $(y)$ solamente?                 |
| $(D_{\mathbf u}f)$        | ¿Qué pasa si me muevo en la dirección $(\mathbf u)$? |
| $(\nabla f)$              | ¿Cuál es la dirección de máximo crecimiento?         |
| $(\|\nabla f\|)$          | ¿Cuál es la máxima tasa de crecimiento?              |

Y la relación fundamental:

$$
\boxed{
D_{\mathbf u}f=\nabla f\cdot\mathbf u
}
$$

# 31. Una forma de recordarlo

Imagina que estás sobre una montaña.

### Derivada parcial

Miras qué ocurre si caminas:

**Este-Oeste**

$$
f_x
$$

o:

**Norte-Sur**

$$
f_y
$$

### Derivada direccional

Dices:

> "Quiero caminar exactamente hacia allá."

$$
D_{\mathbf u}f
$$

### Gradiente

Preguntas:

> "¿En qué dirección debo caminar para subir lo máximo posible?"

$$
\nabla f
$$

Y:

> "¿Qué tan rápido puedo subir?"

$$
\|\nabla f\|
$$

$$

f_x,\quad f_y.
$$

### Nivel 2

Calcular:

$$
\nabla f.
$$

### Nivel 3

Evaluar:

$$
\nabla f(x_0,y_0).
$$

### Nivel 4

Dado:

$$
\mathbf v
$$

normalizarlo:

$$
\mathbf u=\frac{\mathbf v}{\|\mathbf v\|}.
$$

### Nivel 5

Calcular:

$$
D_{\mathbf u}f
=
\nabla f\cdot\mathbf u.
$$

### Nivel 6

Determinar:

$$
\text{máximo crecimiento}

\|\nabla f\|.
$$

### Nivel 7

Determinar la dirección:

$$
\mathbf u_{\max}
=
\frac{\nabla f}{\|\nabla f\|}.
$$

### Nivel 8

Determinar la dirección de máximo decrecimiento:

$$
-\frac{\nabla f}{\|\nabla f\|}.
$$

### Nivel 9

Entender:

\nabla f\perp\text{curvas de nivel}.
