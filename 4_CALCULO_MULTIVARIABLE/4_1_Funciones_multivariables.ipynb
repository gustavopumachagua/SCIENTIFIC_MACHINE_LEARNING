# 1. ¿Qué es una función multivariable?

En cálculo de una variable tenemos:

$$
y=f(x)
$$

Una entrada:

$$
x\in\mathbb R
$$

y una salida:

$$
y\in\mathbb R.
$$

Por ejemplo:

$$
f(x)=x^2+3x.
$$


Con varias variables:

$$
\boxed{z=f(x,y)}
$$

tenemos dos entradas:

$$
(x,y)
$$

y una salida:

$$
z.
$$

Por ejemplo:

$$
\boxed{
f(x,y)=x^2+3xy+y^2
}
$$

Ahora cada punto:

$$
(x,y)
$$

produce un número:

$$
z=f(x,y).
$$

# 2. Ejemplo de vida diaria

Supongamos que quieres estimar el precio de una vivienda.

El precio puede depender de:

* área $(A)$
* número de habitaciones $(R)$
* distancia al centro $(D)$
* antigüedad $(E)$
* piso $(P)$

Entonces podemos escribir:

$$
\boxed{
Precio=f(A,R,D,E,P)
}
$$

Por ejemplo:

$$
P=f(120,3,5,10,4).
$$

Ya no tenemos una función de una variable.

Tenemos:

$$
f:\mathbb R^5\rightarrow\mathbb R.
$$

Esto es exactamente el tipo de estructura que aparece en Machine Learning.

# 3. Ejemplo científico

La temperatura de una habitación depende de:

$$
x,y,z,t
$$

por lo que:

$$
\boxed{
T=T(x,y,z,t)
}
$$

donde:

* $(x)$: posición horizontal
* $(y)$: posición horizontal
* $(z)$: altura
* $(t)$: tiempo

Esto significa que para cada instante tenemos un campo:

$$
T(x,y,z).
$$

Y conforme pasa el tiempo:

$$
T(x,y,z,t).
$$

Esto nos lleva directamente a las **PDE**.

# 5. Graficar una función de dos variables

Una función:

$$
z=f(x,y)
$$

puede visualizarse como una superficie tridimensional.

Consideremos:

$$
\boxed{
f(x,y)=x^2+y^2
}
$$

Es un paraboloide.

Código Python:

````python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-5, 5, 200)
y = np.linspace(-5, 5, 200)

X, Y = np.meshgrid(x, y)

Z = X**2 + Y**2

fig = plt.figure(figsize=(9, 7))

ax = fig.add_subplot(111, projection="3d")

ax.plot_surface(
    X,
    Y,
    Z
)

ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_zlabel("f(x,y)")

ax.set_title(r"$f(x,y)=x^2+y^2$")

plt.show()
````

# 7. Curvas de nivel

Una herramienta fundamental.

Supongamos:

$$
f(x,y)=x^2+y^2.
$$

Fijamos:

$$
f(x,y)=c.
$$

Entonces:

$$
x^2+y^2=c.
$$

Estas son circunferencias.

Por ejemplo:

$$
c=1,4,9,16.
$$

Cada curva representa puntos con el mismo valor de $(f)$.

## Python

````python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-5, 5, 300)
y = np.linspace(-5, 5, 300)

X, Y = np.meshgrid(x, y)

Z = X**2 + Y**2

plt.figure(figsize=(8, 7))

contours = plt.contour(
    X,
    Y,
    Z,
    levels=15
)

plt.clabel(contours)

plt.xlabel("x")
plt.ylabel("y")
plt.title(r"Curvas de nivel de $f(x,y)=x^2+y^2$")
plt.axis("equal")
plt.grid()

plt.show()
````

# 9. Derivadas parciales

Ahora llegamos a uno de los conceptos más importantes.

Consideremos:

$$
f(x,y)=x^2+3xy+y^2.
$$

Queremos saber cómo cambia $(f)$ respecto a $(x)$.

Tratamos $(y)$ como constante.

Entonces:

$$
\frac{\partial f}{\partial x}
=
2x+3y.
$$

Escribimos:

$$
\boxed{
f_x=2x+3y
}
$$


Ahora respecto a $(y)$:

Tratamos $(x)$ como constante:

$$
\boxed{
f_y=3x+2y
}
$$

# 10. ¿Qué significa físicamente?

Supongamos:

$$
T(x,y)
$$

es temperatura.

Entonces:

$$
\frac{\partial T}{\partial x}
$$

significa:

> ¿Cómo cambia la temperatura si me desplazo en dirección $(x)$, manteniendo $(y)$ fijo?

Mientras:

$$
\frac{\partial T}{\partial y}
$$

significa:

> ¿Cómo cambia la temperatura si me desplazo en dirección $(y)$, manteniendo $(x)$ fijo?

Esto es extremadamente importante para física.

# 11. Ejemplo difícil

Consideremos:

$$
T(x,y)
=
100
-2x^2
-3y^2
+4xy
+10x
-8y.
$$

Calculamos:

$$
\frac{\partial T}{\partial x}
=
-4x+4y+10
$$

y:

$$
\frac{\partial T}{\partial y}
=
-6y+4x-8.
$$

Supongamos:

$$
(x,y)=(2,1).
$$

Entonces:

$$
T_x(2,1)
=
-8+4+10
=
6.
$$

Y:

$$
T_y(2,1)
=
-6+8-8
=
-6.
$$

Interpretación:

* movernos en $(x)$ aumenta $(T)$ inicialmente
* movernos en $(y)$ disminuye $(T)$ inicialmente.

# 12. Python: campo y gradientes

Podemos visualizar esta superficie y posteriormente el gradiente.

````python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-4, 4, 200)
y = np.linspace(-4, 4, 200)

X, Y = np.meshgrid(x, y)

T = (
    100
    - 2*X**2
    - 3*Y**2
    + 4*X*Y
    + 10*X
    - 8*Y
)

Tx = -4*X + 4*Y + 10
Ty = 4*X - 6*Y - 8

plt.figure(figsize=(9, 7))

plt.contourf(X, Y, T, levels=30)

plt.colorbar(label="Temperatura")

plt.quiver(
    X[::10, ::10],
    Y[::10, ::10],
    Tx[::10, ::10],
    Ty[::10, ::10]
)

plt.xlabel("x")
plt.ylabel("y")
plt.title("Campo escalar y gradiente de temperatura")

plt.show()
````

# 13. Gradiente

El gradiente reúne todas las derivadas parciales:

$$
\boxed{
\nabla f
=
\begin{bmatrix}
\frac{\partial f}{\partial x}\\
\frac{\partial f}{\partial y}
\end{bmatrix}
}
$$

Para tres variables:

$$
\boxed{
\nabla f=
\begin{bmatrix}
f_x\\
f_y\\
f_z
\end{bmatrix}
}
$$

# 14. ¿Qué significa el gradiente?

Esta propiedad debes aprenderla muy bien:

$$
\boxed{
\nabla f
\text{ apunta hacia la dirección de máximo crecimiento de }f
}
$$

Además:

$$
\boxed{
\|\nabla f\|
\text{ indica la tasa máxima de crecimiento}
}
$$

Por ejemplo, si:

$$
\nabla T=(3,4)
$$

entonces:

$$
\|\nabla T\|
=
5.
$$

La temperatura aumenta más rápidamente en la dirección:

$$
\frac{1}{5}(3,4)
=
\left(\frac35,\frac45\right).
$$

# 15. Relación entre gradiente y curvas de nivel

Existe una propiedad importantísima:

$$
\boxed{
\nabla f
\perp
\text{curvas de nivel}
}
$$

Es decir, si:

$$
f(x,y)=c
$$

describe una curva de nivel, el gradiente apunta perpendicularmente a esa curva.

Esto aparece constantemente en:

* optimización
* fluidos
* electromagnetismo
* transferencia de calor
* Machine Learning.

# 16. Derivadas parciales de segundo orden

Ahora:

$$
f_x
$$

puede volver a derivarse.

Tenemos:

$$
f(x,y)=x^2+3xy+y^2.
$$

Primero:

$$
f_x=2x+3y
$$

Entonces:

$$
f_{xx}=2.
$$

También:

$$
f_y=3x+2y
$$

y:

$$
f_{yy}=2.
$$

Las derivadas cruzadas:

$$
f_{xy}=3
$$

y:

$$
f_{yx}=3.
$$

En condiciones suficientemente suaves:

$$
\boxed{
f_{xy}=f_{yx}
}
$$

Esto se relaciona con el teorema de Clairaut.

# 17. Hessiano

Todas las segundas derivadas pueden organizarse en una matriz:

$$
\boxed{
H=
\nabla^2 f
=
\begin{bmatrix}
f_{xx}&f_{xy}\\
f_{yx}&f_{yy}
\end{bmatrix}
}
$$

Para:

$$
f=x^2+3xy+y^2
$$

tenemos:

$$
\boxed{
H=
\begin{bmatrix}
2&3\\
3&2
\end{bmatrix}
}
$$

El Hessiano será fundamental posteriormente para:

* optimización
* Newton
* estabilidad
* análisis de convexidad
* entrenamiento de modelos
* problemas inversos
* SciML.

# 18. Máximos y mínimos

Para una función:

$$
f(x,y)
$$

un punto crítico ocurre cuando:

$$
\boxed{
\nabla f=0
}
$$

es decir:

$$
f_x=0
$$

y:

$$
f_y=0.
$$

# 19. Ejemplo difícil

Consideremos:

$$
f(x,y)
=
x^2+2y^2-4x-8y+xy.
$$

Calculamos:

$$
f_x=2x+y-4
$$

$$
f_y=4y+x-8.
$$

Igualamos a cero:

$$
2x+y=4
$$

$$
x+4y=8.
$$

Resolviendo:

$$
x=0
$$

$$
y=4.
$$

Entonces:

$$
(0,4)
$$

es un punto crítico.

# 20. Clasificación mediante Hessiano

Tenemos:

$$
H=
\begin{bmatrix}
2&1\\
1&4
\end{bmatrix}.
$$

Su determinante:

$$
\det(H)=8-1=7>0
$$

y:

$$
f_{xx}=2>0.
$$

Por tanto:

$$
\boxed{
(0,4)\text{ es un mínimo local}
}
$$

De hecho, en este caso es un mínimo global porque la función es estrictamente convexa.

# 21. Aplicación a Machine Learning

Supongamos una función de pérdida:

$$
L(w_1,w_2).
$$

Tenemos:

$$
\boxed{
\nabla L
}
$$

que indica cómo modificar los parámetros.

Gradient descent:

$$
\boxed{
w_{k+1}
=
w_k-\eta\nabla L(w_k)
}
$$

Aquí las funciones multivariables dejan de ser una parte abstracta del cálculo.

Son directamente el mecanismo de entrenamiento de ML.

# 22. Ejemplo visual de Gradient Descent

Consideremos:

$$
L(x,y)
=
x^2+3y^2.
$$

Entonces:

$$
\nabla L=
\begin{bmatrix}
2x\\
6y
\end{bmatrix}.
$$

Gradient descent:

$$
x_{k+1}=x_k-2\eta x_k
$$

$$
y_{k+1}=y_k-6\eta y_k.
$$

Podemos visualizar cómo el algoritmo desciende hacia:

$$
(0,0).
$$

````python
import numpy as np
import matplotlib.pyplot as plt

def f(x, y):
    return x**2 + 3*y**2

def grad(x, y):
    return np.array([
        2*x,
        6*y
    ])

eta = 0.08

point = np.array([4.0, 4.0])

trajectory = [point.copy()]

for _ in range(30):

    point = point - eta * grad(
        point[0],
        point[1]
    )

    trajectory.append(point.copy())

trajectory = np.array(trajectory)

x = np.linspace(-5, 5, 300)
y = np.linspace(-5, 5, 300)

X, Y = np.meshgrid(x, y)

Z = f(X, Y)

plt.figure(figsize=(9, 7))

plt.contour(
    X,
    Y,
    Z,
    levels=20
)

plt.plot(
    trajectory[:, 0],
    trajectory[:, 1],
    "o-",
    label="Gradient Descent"
)

plt.xlabel("x")
plt.ylabel("y")
plt.title("Descenso por gradiente sobre una función multivariable")
plt.legend()
plt.grid()

plt.show()
````

# 23. Regla de la cadena multivariable

Ahora supongamos:

$$
z=f(x,y)
$$

pero:

$$
x=x(t)
$$

$$
y=y(t).
$$

Entonces:

$$
\boxed{
\frac{dz}{dt}
=
\frac{\partial f}{\partial x}
\frac{dx}{dt}
+
\frac{\partial f}{\partial y}
\frac{dy}{dt}
}
$$

Esta fórmula será importantísima para ODE y física.

# 24. Ejemplo

Supongamos:

$$
f(x,y)=x^2+y^2
$$

y:

$$
x(t)=t^2
$$

$$
y(t)=e^t.
$$

Entonces:

$$
z(t)=t^4+e^{2t}.
$$

Pero usando regla de la cadena:

$$
f_x=2x
$$

$$
f_y=2y.
$$

Además:

$$
x'=2t
$$

$$
y'=e^t.
$$

Entonces:

$$
\frac{dz}{dt}
=
2x(2t)+2y(e^t)
$$

sustituyendo:

$$
\boxed{
\frac{dz}{dt}
=
4t^3+2e^{2t}
}
$$

# 25. Derivada direccional

El gradiente permite saber cómo cambia $(f)$ en cualquier dirección.

Si:

$$
\mathbf u
$$

es un vector unitario, entonces:

$$
\boxed{
D_{\mathbf u}f
=
\nabla f\cdot\mathbf u
}
$$

## Ejemplo

Supongamos:

$$
f(x,y)=x^2+xy+y^2.
$$

En:

$$
(1,2)
$$

tenemos:

$$
\nabla f
=
\begin{bmatrix}
2x+y\\
x+2y
\end{bmatrix}
$$

por tanto:

$$
\nabla f(1,2)
=
\begin{bmatrix}
4\\
5
\end{bmatrix}.
$$

Queremos movernos en la dirección:

$$
v=(3,4).
$$

Primero normalizamos:

$$
u=
\frac{(3,4)}{5}
=
\left(\frac35,\frac45\right).
$$

Entonces:

$$
D_uf
=
(4,5)\cdot
\left(\frac35,\frac45\right)
$$

$$
=
\frac{12}{5}+\frac{20}{5}
$$

$$
\boxed{
D_uf=6.4
}
$$

# 26. Funciones vectoriales

Hasta ahora:

$$
f:\mathbb R^n\rightarrow\mathbb R.
$$

Pero en SciML también tendremos:

$$
\boxed{
\mathbf F:\mathbb R^n\rightarrow\mathbb R^m
}
$$

Por ejemplo:

$$
\mathbf F(x,y)
=
\begin{bmatrix}
x^2+y\\
xy\\
e^x
\end{bmatrix}.
$$

Tenemos dos entradas:

$$
(x,y)
$$

y tres salidas.

# 27. Jacobiano

El Jacobiano organiza todas las derivadas parciales:

$$
\boxed{
J=
\begin{bmatrix}
\frac{\partial F_1}{\partial x_1}
&
\cdots
&
\frac{\partial F_1}{\partial x_n}
\\
\vdots
&
\ddots
&
\vdots
\\
\frac{\partial F_m}{\partial x_1}
&
\cdots
&
\frac{\partial F_m}{\partial x_n}
\end{bmatrix}
}
$$

Si:

$$
\mathbf F:
\mathbb R^2\rightarrow\mathbb R^3
$$

entonces:

$$
J\in\mathbb R^{3\times2}.
$$

# 28. Ejemplo difícil de Jacobiano

Consideremos:

$$
F_1(x,y,z)=x^2+yz
$$

$$
F_2(x,y,z)=e^x+\sin(yz)
$$

$$
F_3(x,y,z)=xyz.
$$

Entonces:

$$
\frac{\partial F_1}{\partial x}=2x
$$

$$
\frac{\partial F_1}{\partial y}=z
$$

$$
\frac{\partial F_1}{\partial z}=y
$$

Para $(F_2)$:

$$
\frac{\partial F_2}{\partial x}=e^x
$$

$$
\frac{\partial F_2}{\partial y}
=
z\cos(yz)
$$

$$
\frac{\partial F_2}{\partial z}
=
y\cos(yz)
$$

Para $(F_3)$:

$$
\frac{\partial F_3}{\partial x}=yz
$$

$$
\frac{\partial F_3}{\partial y}=xz
$$

$$
\frac{\partial F_3}{\partial z}=xy.
$$

Por tanto:

$$
\boxed{
J=
\begin{bmatrix}
2x&z&y\\
e^x&z\cos(yz)&y\cos(yz)\\
yz&xz&xy
\end{bmatrix}
}
$$

# 29. ¿Por qué el Jacobiano es tan importante en SciML?

Porque una ODE puede escribirse:

$$
\boxed{
\frac{d\mathbf u}{dt}
=
\mathbf F(\mathbf u,t)
}
$$

donde:

$$
\mathbf u\in\mathbb R^n.
$$

El Jacobiano:

$$
\boxed{
J_F=
\frac{\partial\mathbf F}{\partial\mathbf u}
}
$$

describe cómo cambia el sistema cuando modificamos ligeramente $(\mathbf u)$.

Aparece en:

* Newton
* ODE solvers
* sistemas rígidos
* estabilidad
* sensibilidad
* optimización
* adjoint methods
* differentiable programming.

# 30. Linealización

Cerca de un punto:

$$
x_0
$$

podemos aproximar:

$$
\boxed{
f(x)
\approx
f(x_0)
+
\nabla f(x_0)^T(x-x_0)
}
$$

Para una función vectorial:

$$
\boxed{
F(x)
\approx
F(x_0)
+
J(x_0)(x-x_0)
}
$$

Esto es fundamental.

Una función complicada localmente puede comportarse aproximadamente como una función lineal.

# 31. Aplicación física: temperatura

Supongamos:

$$
T(x,y,z,t).
$$

El gradiente espacial es:

$$
\boxed{
\nabla T=
\begin{bmatrix}
T_x\\
T_y\\
T_z
\end{bmatrix}
}
$$

y la derivada temporal:

$$
T_t.
$$

En transferencia de calor aparece:

$$
\boxed{
\frac{\partial T}{\partial t}
=
\alpha\nabla^2T
}
$$

donde:

$$
\boxed{
\nabla^2T
=
T_{xx}+T_{yy}+T_{zz}
}
$$

es el **Laplaciano**.

Aquí puedes ver cómo el cálculo multivariable desemboca directamente en una PDE.

# 32. Laplaciano

Para:

$$
f(x,y)
$$

tenemos:

$$
\boxed{
\nabla^2f
=
f_{xx}+f_{yy}
}
$$

Para tres dimensiones:

$$
\boxed{
\nabla^2f
=
f_{xx}+f_{yy}+f_{zz}
}
$$

## Ejemplo

$$
f(x,y)=x^3+2x^2y+5y^2.
$$

Tenemos:

$$
f_{xx}=6x+4y
$$

y:

$$
f_{yy}=10.
$$

Por tanto:

$$
\boxed{
\nabla^2f=6x+4y+10
}
$$

# 34. Conexión con una red neuronal

Una red neuronal puede verse como una composición:

$$
f(x)
=
W_3
\sigma
\left(
W_2
\sigma(W_1x+b_1)
+b_2
\right)
+b_3.
$$

Esto es una función multivariable extremadamente compleja.

Sus parámetros:

$$
\theta=
(W_1,W_2,W_3,b_1,b_2,b_3)
$$

pueden ser cientos, millones o miles de millones.

La pérdida:

$$
L(\theta)
$$

es una función multivariable.

Entrenar la red significa resolver aproximadamente:

$$
\boxed{
\min_\theta L(\theta)
}
$$

y necesitamos:

$$
\nabla_\theta L.
$$

Ahí aparecen:

* derivadas parciales
* regla de la cadena
* Jacobianos
* Hessianos
* cálculo multivariable.

# 35. Conexión todavía más profunda: PINNs

En Physics-Informed Neural Networks tenemos una red:

$$
u_\theta(x,t).
$$

La red intenta aproximar una solución de una PDE.

Por ejemplo:

$$
u_t-\alpha u_{xx}=0.
$$

Construimos el residual:

$$
\boxed{
r_\theta(x,t)
=
\frac{\partial u_\theta}{\partial t}
-
\alpha
\frac{\partial^2u_\theta}{\partial x^2}
}
$$

$$

La función de pérdida puede ser:

$$

=

$$

Para calcularla necesitamos derivadas de:

$$

# 37. Fórmulas que debes tener dominadas

### Gradiente

$$
\boxed{
\nabla f=
\begin{bmatrix}
f_{x_1}\\
\vdots\\
f_{x_n}
\end{bmatrix}
}
$$

### Derivada direccional

$$
\boxed{
D_{\mathbf u}f=\nabla f\cdot\mathbf u
}
$$


### Hessiano

$$
\boxed{
H=
\nabla^2f
}
$$

### Jacobiano

$$
\boxed{
J_{ij}
=
\frac{\partial F_i}{\partial x_j}
}
$$

### Taylor multivariable de primer orden

$$
\boxed{
f(x)
\approx
f(x_0)
+
\nabla f(x_0)^T(x-x_0)
}
$$

### Taylor de segundo orden

$$
\boxed{
f(x)
\approx
f(x_0)
+
\nabla f(x_0)^T\Delta x
+
\frac12
\Delta x^TH(x_0)\Delta x
}
$$

### Laplaciano

$$
\boxed{
\nabla^2f
=
\sum_i
\frac{\partial^2f}{\partial x_i^2}
}
$$
